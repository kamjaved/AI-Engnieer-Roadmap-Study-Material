### A2. Password Sanitizer — *strings* · Easy

> "A single line of space-separated passwords … return the valid ones, also space-separated and in the same order. A password passed if it had **at least 5 characters**, was **not made up only of letters**, and was **not made up only of digits**."

| Input | Output |
|---|---|
| `"abc123 abcd 123456 Passw0rd abcdef"` | `"abc123 Passw0rd"` |
| `"a!b@c# 12345 hello"` | `"a!b@c#"` |
| `""` | `""` |

**Assumptions to check against the real statement:** (1) symbol-only passwords like `"!!!!!"` count as valid (not all letters, not all digits); (2) empty result → empty string (some versions want `"NONE"` or `-1` — read the output spec); (3) multiple spaces between tokens — `str.split()` with no argument handles that.

<details><summary>Solution — O(T) time, T = total characters</summary>


```python
MIN_PASS_LEN=5

def sanatize_password(passwords:str)->str:
  valid= [p for p in passwords.split() if len(p)>=MIN_PASS_LEN and not p.isalpha() and not p.isdigit()]
  return " ".join(valid)

print(sanatize_password("abc123 abcd 123456 Passw0rd abcdef"))
```

### A4. Majority Element — *arrays / Boyer–Moore* · Easy

> Find the value that appears **more than half** the time. Example `[2,2,1,1,1,2,2]` → `2`. Expected O(n) time, O(1) space.

```python
def majority_element(nums: [int]) -> int:
    candidate, count = 0, 0
    for n in nums:
        if count == 0:
            candidate = n

        count +=1 if n== candidate else -1
    return  candidate

print(majority_element([2,2,1,1,1,2,2]))
print(majority_element([5,5,5,2,2,1,1,1,2,2,5,5,5,5,5]))
```

```python
def reverse_words_order_and_swap_cases(sentence:str):
  #  reversed_words = [word.swapcase() for word in sentence.split(" ")[::-1]]
  #  return " ".join(reversed_words)
  reversed_words= [word.swapcase() for word in sentence.split()[::-1]]
  return " ".join(reversed_words)

print(reverse_words_order_and_swap_cases('aWESOME is cODING'))


#  ANOTHER METHOD

def reverse_words(sentence: str) -> str:
    return " ".join(reversed(sentence.split()))

def reverse_words_swap_case(sentence: str) -> str:
    return reverse_words(sentence).swapcase()

print(reverse_words_swap_case('aWESOME is cODING'))
```

```python
def remove_duplicates_seen(array:list[str])->list[str]:
  seen=set()
  result=[]
  for item in array:
    if item not in seen:
      seen.add(item)
      result.append(item)
  return result

print(remove_duplicates_seen(["doc3", "doc1", "doc2", "doc2", "doc1"]))


def remove_duplicates_seen_v2(array:list[str])->list[str]:
  return list(dict.fromkeys(array))

print(remove_duplicates_seen_v2(["doc3", "doc1", "doc2", "doc2", "doc1"]))
print(remove_duplicates_seen_v2([1, 1, 2, 3, 2, 1]))
```

```python
import string


def top_n_words(sentence: str) -> dict[str, int]:
  stripped_text= sentence.translate(str.maketrans("", "", string.punctuation))
  words= stripped_text.lower().split()

  result={}

  for word in words:
      if word in result:
        result[word]+=1
      else:
        result[word]=1
  # for key, value in result.items():
  #   #  print(f"KEY: '{key}' VALUE: '{value}' ")
  return result


print(top_n_words("The cat sat. The cat ran! the DOG sat"))

print("Hello World".strip("w"))
```

```python
def two_sum(nums: list[int], target: int) -> list[int]:
  seen:dict[int, int]={}

  for index,num in enumerate(nums):
    complement=target-num
    if complement in seen:
      return [seen[complement], index]
    seen[num]=index

  return []



print(two_sum([2, 7, 11, 15], 9))
```

```python
def is_anagram_sorted_v2(a: str, b: str) -> bool:
  text_a= [ch.lower() for ch in a if ch.isalnum()]
  text_b= [ch.lower() for ch in b if ch.isalnum()]

  return sorted(text_a) == sorted(text_b)

print(is_anagram_sorted_v2("Listen", "Silent"))
print(is_anagram_sorted_v2("elbow", "below"))
print(is_anagram_sorted_v2("Debit Card", "Bad Credit"))
print(is_anagram_sorted_v2("hello",  "world" ))
print(is_anagram_sorted_v2("abcd",  "abded" ))
```

```python

def fib_series(n:int)->list[int]:
  a,b=-0,1
  series=[]
  for _ in range(n):
    series.append(a)
    a, b = b, a + b
  return series


def fib_nth(n:int)->int:
  a,b= 0,1
  for _ in range(n):
    a,b=b, a+b
  return a


def is_fibonacci(n: int) -> bool:

  def is_square(x:int)->bool:
    return x>=0 and issqrt(x)**2 ==x
  
  return n>=0 and (is_square(5 *n*n) +4 or is_square(5 * n * n - 4))


print(fib_series(13))  
print(fib_nth(7)) 
print(is_fibonacci(21), is_fibonacci(22))  # True False 
```

```python
from collections import Counter


def character_frequency(s:str)->dict[str, int]:

  seen:dict[str, int]={}
  for char in s:
    seen[char]= seen.get(char, 0)+1
  return seen

def character_frequency_v2(s:str)->dict[str, int]:
    return dict(Counter(s).most_common())


def first_unique(s: str) -> str | None:
    counts = Counter(s)
    return next((ch for ch in s if counts[ch] == 1), None)
   

print(character_frequency("hellllo"))
print(character_frequency_v2("hhellllo"))
print(first_unique("swiss"))



```

```python
def is_palindrome(s: str) -> bool:
    cleaned=[char.lower() for char in s if char.isalpha()]
    return cleaned==cleaned[::-1]

print(is_palindrome("radar"))
```

```python
# FLATTEN-NESTED-LEVEL 

def flatten_deep(nested:list)->list:
  result=[]
  for item in nested:
    if isinstance(item, list):
       result.extend(flatten_deep(item))
    else:
      result.append(item)
  return result

print(flatten_deep([1, [2, [3, [4, 5]], 6], 7]))
```

```python
import heapq

names = ["asha", "ravi", "meena"]
scores = [91, 78, 85]

pairs = list(zip(names, scores, strict=True))   # strict= raises if lengths differ (3.10+)
lookup = dict(zip(names, scores, strict=True))               # {'asha': 91, 'ravi': 78, 'meena': 85}
pass_list = [n for n, s in zip(names, scores) if s >= 80]   # merge-by-condition
merged_sorted = list(heapq.merge([1, 4, 9], [2, 3, 10]))    # [1, 2, 3, 4, 9, 10]
names_back, scores_back = zip(*pairs)            # "unzip"

print(pairs)
print(lookup)
print(pass_list)
print(merged_sorted)
print(names_back)
print(scores_back)
```
