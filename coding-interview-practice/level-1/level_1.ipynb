**Question 1**

Write a function `count_frequency(nums)` that takes a list of integers and returns a dictionary mapping each integer to the number of times it appears in the list. Do not use `collections.Counter` — implement the counting logic yourself.

**Example**

```
Input:  [1, 2, 2, 3, 3, 3, 4]
Output: {1: 1, 2: 2, 3: 3, 4: 1}
```

**What's expected**

Write clean, working Python. Order of keys in the output doesn't matter. Handle an empty list input correctly.

**What this tests**

Baseline fluency with dictionaries — key existence checks, default values, and iteration — before I introduce `Counter`/`defaultdict` shortcuts later. This is the most repeated warm-up pattern across interview prep sources, and it's also the foundation for anagram/first-non-repeating-character problems coming up in Tier 3.

Go ahead and submit your solution.

````python

def count_frequency(nums:list[int])->dict[int,int]:
    result:dict[int,int]={}
    for value in nums:
     if value in result:
          result[value]+=1
     else:
       result[value]=1
    print(result)
             

count_frequency([1, 2, 2, 3, 3, 3, 4])

# Input:  [1, 2, 2, 3, 3, 3, 4]
# Output: {1: 1, 2: 2, 3: 3, 4: 1}
````

### The Same things can be achived using `Counter` Method from collection package

````python

from collections import Counter
def count_frequency(nums:list[int])->dict[int, int]:
  return dict(Counter(nums))

result = count_frequency([1, 2, 2, 3, 3, 3, 4])
print(result)

 
````

## Q1. Reverse a string, then reverse the word order of a sentence

**Difficulty:** Beginner

**Problem statement**
Write two small functions. The first reverses the characters of a string. The second reverses
the *order of words* in a sentence, keeping each word itself intact. Do not use any external
library.

**Example**
```
Input:  "genai"
Output: "ianeg"

Input:  "I build production AI systems"
Output: "systems AI production build I"
```

**Expected task**
Implement both. Then answer out loud: what happens with multiple spaces between words
(`"a    b"`), and with leading/trailing whitespace?

**Concept / pattern tested**
String slicing with a negative step, `str.split()`, `str.join()`, immutability of strings.

**Why it's asked**
This is the most common opening warm-up in Python screens. Interviewers use it to check
whether you reach for `[::-1]` and `split()/join()` naturally, or start writing a manual
index loop like a C programmer. In AI work you touch text constantly — clean string handling
is the base layer under every prompt builder, parser and chunker.

````python
def reverseCharacter(char:str)->str:
    # return "".join(reversed(char))
    return char[::-1]

result = reverseCharacter("genai")
print (result)


def reverseWord(words:str)->str:
     return " ".join(words.split()[::-1])

result2 = reverseWord("I build production AI   systems")
print (result2)
````

## Q2. Count character frequency using a plain dictionary

**Difficulty:** Beginner

**Problem statement**
Given a string, return a dictionary mapping each character to the number of times it appears.
Ignore case. **Do not use `collections.Counter` for this one** — write the counting loop by hand.

**Example**
```
Input:  "Hello"
Output: {'h': 1, 'e': 1, 'l': 2, 'o': 1}
```

**Expected task**
Solve it with `dict.get()` and then again with `dict.setdefault()`. Explain the difference
between the two, and why `dict[key] += 1` fails on a missing key.

**Concept / pattern tested**
Dictionary as a counter, `get()` vs `setdefault()` vs `KeyError`, O(1) average hash lookup.

**Why it's asked**
Frequency counting is the single most reused pattern in Python interviews — anagrams,
first-unique-character, top-K, word counts and duplicate detection are all this loop in
disguise. Interviewers deliberately ban `Counter` first, then ask you to redo it with `Counter`,
to see if you understand what the library is doing for you.


````python

def frequency_counter(word: str)->dict[str, int]:
   listWord= list(word.lower())
   result = {}
   for value in listWord:
      if value in result:
           result[value]+=1
      else:
          result[value]=1

   return result

result= frequency_counter("Hello")
print(result)



    
def collection_counter(word:str)->dict['str', int]:
    return dict(Counter(word.lower()))

result3= collection_counter("Helello")
print(result3)


# RECOMMNEDED CLEAN VERSION
# listWord = list(word) is redundant in 1st attempt. Strings are already iterable — for ch in word

def char_frequency(text: str) -> dict[str, int]:
    """Count characters, case-insensitively. Manual loop, no Counter."""
    counts: dict[str, int] = {}
    for char in text.lower():          # normalise once, up front — not inside the loop
        counts[char] = counts.get(char, 0) + 1
    return counts

result2= char_frequency("HHeheelloo")
print(result2)
````

## Q3. Find the second largest number in a list without sorting

**Difficulty:** Beginner

**Problem statement**
Given a list of integers, return the second largest **distinct** value. You may not call
`sorted()`, `list.sort()`, `max()` more than once conceptually, or convert to a set to cheat
the logic — do it in a single pass with two variables.

**Example**
```
Input:  [10, 5, 22, 22, 7]
Output: 22 is largest, so answer = 10

Input:  [4, 4, 4]
Output: None  (no second distinct value)
```

**Expected task**
Handle these edge cases explicitly: empty list, single element, all elements equal,
negative numbers.

**Concept / pattern tested**
Single-pass scanning, tracking two running values, edge-case discipline, why
`float('-inf')` is a safer initial value than `0`.

**Why it's asked**
It is the classic "can you think without reaching for a built-in" question. Most candidates
sort in O(n log n); the interviewer wants the O(n) single pass. The real signal is whether you
volunteer the edge cases before being asked — that is what separates 2–3 years of experience
from a fresher.

````python

# SOLUTION WITH using Sorted, set Not accroding to Problem Statement But if you have freedom and only means to result then solution is valid
def find_second_largest(numbers: list[int]) -> str | None:
    # Remove duplicates and sort in ascending order
    unique_sorted = sorted(set(numbers))
    
    # If there are fewer than 2 unique numbers, no second largest exists
    if len(unique_sorted) < 2:
        return None
    
    largest_number = unique_sorted[-1]
    second_largest = unique_sorted[-2]
    
    return f"{largest_number} is largest, so answer = {second_largest}"

print(find_second_largest([10, 5, 22, 22, 7]))

print(find_second_largest([4, 4, 4]))


print("--------------------SOLUTION -B---------")
#  SOLUTION AS PER PROBLEM STATEMENT USING 2 POinter

def find_second_largest_v2(numbers: list[int]) -> str | None:
    # Handle base edge cases: empty list or single element
    if not numbers or len(numbers) < 2:
        return None

    # Initialize to negative infinity to support lists with negative numbers
    largest = float("-inf")
    second_largest = float("-inf")

    for num in numbers:
        # Case 1: Found a new strictly larger number
        if num > largest:
            second_largest = largest
            largest = num
        # Case 2: Found a number between largest and second_largest (distinct)
        elif num < largest and num > second_largest:
            second_largest = num

    # If second_largest was never updated, all numbers were identical
    if second_largest == float("-inf"):
        return None

    return f"{largest} is largest, so answer = {second_largest}"

# --- Test Cases ---
print(find_second_largest_v2([10, 5, 22, 22, 7]))  # 22 is largest, so answer = 10
print(find_second_largest_v2([4, 4, 4]))            # None
print(find_second_largest_v2([]))                   # None
print(find_second_largest_v2([7]))                  # None
print(find_second_largest_v2([-10, -5, -22, -3]))   # -3 is largest, so answer = -5
````

## Q4. Remove duplicates from a list while preserving original order

**Difficulty:** Beginner

**Problem statement**
Given a list that may contain duplicates, return a new list with duplicates removed, keeping
the order of first appearance.

**Example**
```
Input:  ["doc3", "doc1", "doc3", "doc2", "doc1"]
Output: ["doc3", "doc1", "doc2"]
```

**Expected task**
Write it with an explicit `seen` set first. Then write the idiomatic one-liner using
`dict.fromkeys()`. Explain why `list(set(items))` is the wrong answer here.

**Concept / pattern tested**
Set membership as O(1) lookup, insertion-ordered dicts (Python 3.7+), the order-vs-speed
trade-off.

**Why it's asked**
Interviewers love this one because the naive answer (`list(set(...))`) is *almost* right and
silently destroys ordering. In a RAG pipeline you dedupe retrieved chunk IDs constantly, and
order there is relevance order — losing it is a real production bug, not a trivia point.


````python
# METHOD-1 (MY SOLUTION)
def rmv_duplicate(array:list[str] | list[int])->list[str] | list[int]:
   result: list[str] | list[int] | list = []


   if len(array)==0:
      result=[]
      return result
   if len(array)==1:
    result=array
    return result

   for item in array:
     if item not in result:
       seen=item
       result.append(seen)
       
   return result

print("------SOLUTION-1------")
print(rmv_duplicate(["doc3", "doc1", "doc2", "doc2", "doc1"]))
print(rmv_duplicate([1,1,2,3,2,1]))
print(rmv_duplicate(["a"]))
print(rmv_duplicate([]))

# METHOD-2 
def remove_duplicates_seen(items: list) -> list:
   seen=set()
   result=[]
   for item in items:
      if item not in seen:
         seen.add(item)
         result.append(item)
   return result

print("------SOLUTION-2------")
print(remove_duplicates_seen(["doc3", "doc1", "doc3", "doc2", "doc1"]))
print(remove_duplicates_seen([1, 1, 2, 3, 2, 1]))
print(remove_duplicates_seen("a"))
print(remove_duplicates_seen([]))

# SOLUTION-3 Short One LIner Modern
# Since Python 3.7+, standard dictionaries are guaranteed to maintain insertion order.
#  dict.fromkeys() builds a dictionary using the list elements as keys (which automatically removes duplicates) 
# and preserves the order in which they were inserted.

def remove_duplicates_fast(items: list) -> list:
    return list(dict.fromkeys(items))

print("------SOLUTION-3------")
print(remove_duplicates_fast(["doc3", "doc1", "doc3", "doc2", "doc1"]))
print(remove_duplicates_fast([1, 1, 2, 3, 2, 1]))
print(remove_duplicates_fast("a"))
print(remove_duplicates_fast([]))

````

## Q5. Build a dictionary from two lists using `zip()`

**Difficulty:** Beginner

**Problem statement**
Given a list of keys and a list of values, build a dictionary pairing them up.
Then extend it: what should happen when the two lists have different lengths?

**Example**
```
keys   = ["model", "temperature", "max_tokens"]
values = ["gpt-4o", 0.2, 512]

Output: {"model": "gpt-4o", "temperature": 0.2, "max_tokens": 512}
```

**Expected task**
Implement with `dict(zip(...))`. Then demonstrate what `zip()` does on unequal lengths, and
show how `zip(..., strict=True)` (Python 3.10+) turns a silent truncation into a loud error.

**Concept / pattern tested**
`zip()` semantics, lazy iterators, `dict()` construction from pairs, fail-fast design.

**Why it's asked**
`zip()` is a top-5 "do you actually write Python" tell. The `strict=True` follow-up is the
modern part most candidates miss — silent truncation of mismatched data is exactly the kind of
bug that corrupts an embedding batch where vectors and IDs get out of alignment.


````python
def combine_key_value(keys:list, value:list)->dict:
  return dict(zip(keys, value, strict=True))


keys   = ["model", "temperature", "max_tokens"]
values = ["gpt-4o", 0.2, 512]

print(combine_key_value(keys, values))
````

## Q6. Use `enumerate()` to find every index of a target value

**Difficulty:** Beginner

**Problem statement**
Given a list and a target value, return a list of **all** indices where the target appears.
Then produce a human-readable numbered listing starting at 1.

**Example**
```
Input:  ["a", "b", "a", "c", "a"], target = "a"
Output: [0, 2, 4]

Numbered listing:
1. a
2. b
3. a
4. c
5. a
```

**Expected task**
Solve with `enumerate()` inside a list comprehension. Use the `start=` parameter for the
numbered listing. Explain why `for i in range(len(items))` is considered unidiomatic.

**Concept / pattern tested**
`enumerate()` including its `start` argument, comprehension with a condition, index-vs-value
iteration.

**Why it's asked**
`list.index()` only returns the first match — interviewers use this to see if you notice that.
`enumerate` also appears in nearly every real loop you'll write over chunks, messages or
retrieved documents where you need both the position and the item.


````python
def find_index_value(items:list[str], target: str)->tuple[list[int], str]:
    index_of_target:list[int] = []
    number_listing:list[str]=["Number Listing:"]
    
    # Use start=1 for 1-based numbering
    for index, value in enumerate(items, start=1):
      if value==target:
        index_of_target.append(index-1) # 0-based index for target match
    
      number_listing.append(f"{index}. {value}")

    formatted_listing = "\n".join(number_listing)
    return index_of_target, formatted_listing

indices, listing= find_index_value(["a", "b", "a", "c", "a"], "a")
print("Output:", indices)
print()
print(listing)

  
````

## Q7. Clean and filter a messy list of strings with a comprehension

**Difficulty:** Beginner

**Problem statement**
You receive a list of raw user-supplied tags. Produce a cleaned list: strip whitespace,
lowercase everything, drop empty/whitespace-only entries, and drop anything shorter than
2 characters.

**Example**
```
Input:  ["  RAG ", "", "LangGraph", "   ", "a", "Vector DB  "]
Output: ["rag", "langgraph", "vector db"]
```

**Expected task**
Write it as a single list comprehension. Then write the same thing as an explicit `for` loop
and say which version you would put in a production codebase and why.

**Concept / pattern tested**
List comprehension with transform + filter, order of operations inside a comprehension,
readability limits of comprehensions.

**Why it's asked**
Comprehensions are the number-one thing interviewers scan for. But the real question underneath
is judgement: a candidate who crams four transformations into one unreadable comprehension is
showing off, not engineering. Say the trade-off out loud.


````python

# SOLUTION 1 (MY SOLUTION)
def clean_tags(tags: list[str]) -> list[str]:
    cleaned_tags = []
    
    for tag in tags:
        cleaned = tag.strip().lower()
        # len >= 2 automatically excludes empty strings and single characters
        if len(cleaned) >= 2:
            cleaned_tags.append(cleaned)
            
    return cleaned_tags

raw_tags = ["  RAG ", "", "LangGraph", "   ", "a", "Vector DB  "]
print('------SOLUTOIN-1---------')

print(clean_tags(raw_tags))


def clean_tags_v2(tags: list[str]) -> list[str]:
    return [ cleaned for tag in tags if len(cleaned := tag.strip().lower()) >= 2 ]

# Test
raw_tags = ["  RAG ", "", "LangGraph", "   ", "a", "Vector DB  "]
print('------SOLUTOIN-2---------')
print(clean_tags_v2(raw_tags))
# Output: ['rag', 'langgraph', 'vector db']
````
