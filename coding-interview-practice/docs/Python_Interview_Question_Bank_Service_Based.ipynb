## Rank 3 — Fibonacci (Series, Nth Term & Membership Check)

**Category:** Coding / Loops & Math · **Difficulty:** Easy · **Evidence [CR]:** TCS, Capgemini, Wipro, Citadel

**Problem Statement**
1. Generate the first `n` Fibonacci numbers starting from `0, 1`.
2. Return the 1-indexed `n`-th Fibonacci number where `fib(1) = 1, fib(2) = 1`.
3. Determine whether a given non-negative integer `n` belongs to the Fibonacci sequence.

**Examples**
- `fib_series(7)` → `[0, 1, 1, 2, 3, 5, 8]`
- `nth_fib(4)` → `3`
- `is_fibonacci(21)` → `True`, `is_fibonacci(22)` → `False`

```python
from math import isqrt

def fib_series(n: int) -> list[int]:
    """Return the first n Fibonacci numbers starting with 0, 1."""
    series: list[int] = []
    a, b = 0, 1
    for _ in range(n):
        series.append(a)
        a, b = b, a + b
    return series

def nth_fib(n: int) -> int:
    """Return the 1-indexed nth Fibonacci number where fib(1) = fib(2) = 1."""
    if n <= 0:
        raise ValueError("n must be >= 1")
    a, b = 1, 1
    for _ in range(n - 1):
        a, b = b, a + b
    return a

def is_fibonacci(n: int) -> bool:
    """n is a Fibonacci number iff 5*n^2 + 4 or 5*n^2 - 4 is a perfect square."""
    def is_square(x: int) -> bool:
        return x >= 0 and isqrt(x) ** 2 == x
    return n >= 0 and (is_square(5 * n * n + 4) or is_square(5 * n * n - 4))

print("Series (7):", fib_series(7))
print("4th Fib (1-indexed):", nth_fib(4))
print("Is 21 Fib?:", is_fibonacci(21), "| Is 22 Fib?:", is_fibonacci(22))
```

### Explanation & Complexity Analysis (Rank 3)

**How We Achieved This Solution**
- **Iterative Tuple Swap (`a, b = b, a + b`):** Each Fibonacci number depends only on the previous two values. Instead of storing the entire sequence or building a recursion tree, we maintain just two rolling state variables `a` and `b`. Python evaluates the right-hand side `(b, a + b)` before unpacking, so no temporary variable is needed.
- **Membership via Binet's Property:** A classic number-theory identity states that a non-negative integer $n$ is a Fibonacci number if and only if $5n^2 + 4$ or $5n^2 - 4$ is a perfect square. Using `math.isqrt(x)` avoids floating-point precision errors that `math.sqrt(x)` suffers from on large integers.

**How Time & Space Complexity Work**
- **`fib_series(n)`:** The loop runs `n` times doing O(1) work per step → **O(n) time**. Storing all `n` values in `series` takes **O(n) space**.
- **`nth_fib(n)`:** Loops `n - 1` times with only two variables `a` and `b` → **O(n) time, O(1) space**.
- **`is_fibonacci(n)`:** Performs a constant number of arithmetic operations and integer square roots → **O(1) time, O(1) space** (ignoring arbitrary-precision big-int cost).

**Comparison with Alternative Approaches**

| Approach | Time Complexity | Space Complexity | Trade-offs / Interview Notes |
|---|---|---|---|
| **Naive Recursion** (`fib(n-1) + fib(n-2)`) | O(2^n) | O(n) call stack | Recomputes the same subproblems exponentially; stalls around `n = 35`. |
| **Recursion + `@functools.cache`** | O(n) | O(n) cache + stack | Easy to write if interviewer asks for recursion, but uses extra memory and hits recursion limit around `n = 1000`. |
| **Iterative Rolling Variables (Optimal)** | **O(n)** | **O(1)** (nth) / **O(n)** (series) | Best balance of readability, speed, and memory; no recursion-depth risk. |
| **Matrix Exponentiation** | O(log n) | O(1) | Fastest for huge `n` (e.g., `n = 10^18`), rarely needed unless explicitly asked. |

## Rank 4 — Character / Element Frequency & First Non-Repeating Character

**Category:** Coding / Dictionaries · **Difficulty:** Easy · **Evidence [CR]:** Accenture, Deloitte

**Problem Statement**
1. Count the occurrences of each character in a string using a hash map (`dict`).
2. Find the first character in a string that appears exactly once (return `None` if none exists).

**Examples**
- `char_frequency("hello")` → `{'h': 1, 'e': 1, 'l': 2, 'o': 1}`
- `first_unique("swiss")` → `'w'`

```python
from collections import Counter

def char_frequency(s: str) -> dict[str, int]:
    """Count character frequencies manually using a dictionary."""
    freq: dict[str, int] = {}
    for ch in s:
        freq[ch] = freq.get(ch, 0) + 1
    return freq

def first_unique(s: str) -> str | None:
    """Return the first non-repeating character, or None."""
    counts = Counter(s)
    return next((ch for ch in s if counts[ch] == 1), None)

print("Frequency:", char_frequency("hello"))
print("Most common:", Counter("hello").most_common(1))
print("First unique in 'swiss':", first_unique("swiss"))
print("First unique in 'aabb':", first_unique("aabb"))
```

### Explanation & Complexity Analysis (Rank 4)

**How We Achieved This Solution**
- **Manual Hash Map (`dict.get`):** We iterate through the string once. For each character `ch`, `freq.get(ch, 0)` looks up its current count in O(1) average time (defaulting to `0` if unseen) and increments it by 1. Interviewers prefer seeing this manual pattern first to confirm you understand hashing before mentioning `collections.Counter`.
- **Two-Pass First Unique:** A single pass cannot know whether a character will repeat later in the string. Pass 1 builds the frequency map (`Counter(s)`), and Pass 2 scans `s` left-to-right returning the first character with count `1`. Since Python 3.7+, `dict` preserves insertion order, so you can even iterate over `counts.items()` in Pass 2.

**How Time & Space Complexity Work**
- **Time — O(n):** Each character is visited once during counting (and at most once more in `first_unique`), with O(1) average hash-table lookup/update per character.
- **Space — O(k):** Where `k` is the number of distinct characters (for ASCII, `k <= 128`, which is effectively O(1); for arbitrary Unicode strings, `k <= n`).

**Comparison with Alternative Approaches**

| Approach | Time Complexity | Space Complexity | Trade-offs / Interview Notes |
|---|---|---|---|
| **Nested loop / `s.count(ch)` in loop** | O(n²) | O(1) | `s.count(ch)` scans the entire string on every iteration—slow on large strings. |
| **Sort first, then group** | O(n log n) | O(n) | Destroys original character order, so it cannot directly find the *first* unique char. |
| **Hash Map (`dict.get` / `Counter`) (Optimal)** | **O(n)** | **O(k)** | Linear time, preserves insertion order, idiomatic Python. |

## Rank 7 — Reverse String, Reverse Words, Swap Case & Palindrome

**Category:** Coding / Strings · **Difficulty:** Easy · **Evidence [CR + PP]:** LTIMindtree, HackerRank

**Problem Statement**
Implement four classic string transformations:
1. Reverse a string (both with slicing and manually without slicing).
2. Reverse the order of words in a sentence (collapsing extra whitespace).
3. Reverse words and swap character cases (`Hello World` → `wORLD hELLO`).
4. Check if a string is a palindrome ignoring case and non-alphanumeric characters.

**Examples**
- `reverse_words("Python is awesome")` → `"awesome is Python"`
- `is_palindrome("A man, a plan, a canal: Panama")` → `True`

```python
def reverse_string(s: str) -> str:
    return s[::-1]

def reverse_string_manual(s: str) -> str:
    """Two-pointer reversal when slicing is disallowed."""
    chars = list(s)
    i, j = 0, len(chars) - 1
    while i < j:
        chars[i], chars[j] = chars[j], chars[i]
        i += 1
        j -= 1
    return "".join(chars)

def reverse_words(sentence: str) -> str:
    return " ".join(reversed(sentence.split()))

def reverse_words_swap_case(sentence: str) -> str:
    return reverse_words(sentence).swapcase()

def is_palindrome(s: str) -> bool:
    cleaned = [c.lower() for c in s if c.isalnum()]
    return cleaned == cleaned[::-1]

def is_palindrome_o1_space(s: str) -> bool:
    """Two-pointer palindrome check using O(1) extra space."""
    i, j = 0, len(s) - 1
    while i < j:
        while i < j and not s[i].isalnum():
            i += 1
        while i < j and not s[j].isalnum():
            j -= 1
        if s[i].lower() != s[j].lower():
            return False
        i += 1
        j -= 1
    return True

print("Reverse:", reverse_string("python"), "| Manual:", reverse_string_manual("python"))
print("Reverse words:", reverse_words("  Python   is  awesome  "))
print("Swap case:", reverse_words_swap_case("Hello World"))
print("Palindrome:", is_palindrome("A man, a plan, a canal: Panama"),
      "| O(1) space:", is_palindrome_o1_space("A man, a plan, a canal: Panama"))
```

### Explanation & Complexity Analysis (Rank 7)

**How We Achieved This Solution**
- **Two-Pointer Reversal:** Strings in Python are immutable, so we convert `s` to a mutable `list[str]`, swap elements from opposite ends (`i` and `j`) moving inward until they meet, and join back with `"".join(chars)`.
- **`sentence.split()` vs `split(" ")`:** Calling `.split()` with no arguments automatically strips leading/trailing whitespace and collapses multiple consecutive spaces into one delimiter—critical for clean word reversal.
- **Palindrome:** Filtering into a list and comparing `cleaned == cleaned[::-1]` is concise; if asked to optimize space, two pointers skip non-alphanumeric characters in place without allocating a cleaned copy.

**How Time & Space Complexity Work**
- **String / Word Reversal:** Every character is read and copied into the output string once → **O(n) time, O(n) space** (since Python strings are immutable, returning a new string always takes O(n) space).
- **Two-Pointer Palindrome:** Each index is visited at most once → **O(n) time, O(1) extra space** (vs. O(n) space for the filtered-list approach).

**Comparison with Alternative Approaches**

| Approach | Time Complexity | Space Complexity | Trade-offs / Interview Notes |
|---|---|---|---|
| **Loop with `res = ch + res` or `res += ch`** | O(n²) | O(n) | Strings are immutable; every concatenation allocates and copies the whole string so far. Avoid! |
| **Slicing `s[::-1]` / `"".join(...)` (Optimal)** | **O(n)** | **O(n)** | Implemented in C; fastest and cleanest in Python. |
| **Two-Pointer on `list(s)`** | **O(n)** | **O(n)** | Use when interviewer bans slicing or `reversed()`. |
| **Two-Pointer Palindrome (in-place scan)** | **O(n)** | **O(1)** | Best space complexity for palindrome verification. |

## Rank 8 — Sort an Array (Without Built-in `sort()`) & Alternate Elements

**Category:** Coding / Algorithms · **Difficulty:** Easy · **Evidence [CR]:** Infosys, Wipro, Cognizant

**Problem Statement**
1. Sort a list of integers in ascending order without using Python's built-in `sort()` or `sorted()`.
2. Return every alternate element from a sorted array (indices 0, 2, 4, ...).

**Examples**
- `merge_sort([5, 2, 9, 1, 5, 6])` → `[1, 2, 5, 5, 6, 9]`
- `sorted_alternates([5, 2, 9, 1, 5, 6])` → `[1, 5, 6]`

```python
def merge_sort(arr: list[int]) -> list[int]:
    """Stable O(n log n) divide-and-conquer sort."""
    if len(arr) <= 1:
        return arr[:]
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    merged: list[int] = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:  # <= preserves stability
            merged.append(left[i])
            i += 1
        else:
            merged.append(right[j])
            j += 1
    merged.extend(left[i:])
    merged.extend(right[j:])
    return merged

def sorted_alternates(arr: list[int]) -> list[int]:
    return sorted(arr)[::2]

data = [5, 2, 9, 1, 5, 6]
print("Merge sort:", merge_sort(data))
print("Sorted alternates:", sorted_alternates(data))
```

### Explanation & Complexity Analysis (Rank 8)

**How We Achieved This Solution**
- **Divide and Conquer (Merge Sort):** Recursively split the array in half until subarrays have length 0 or 1 (which are trivially sorted). Then merge pairs of sorted halves using two pointers `i` and `j`, always picking the smaller front element. Using `<=` keeps equal elements in their original relative order (**stable sort**).
- **Production Note:** Always mention that in real code you'd use `sorted()` (returns a new list) or `list.sort()` (sorts in place, returns `None`), both powered by **Timsort** (hybrid Merge + Insertion sort, O(n) on already-sorted data, O(n log n) worst case).

**How Time & Space Complexity Work**
- **Time — O(n log n):** Splitting the array in half creates `log₂(n)` recursion levels. At each level, merging all subarrays touches every element once (`O(n)` work per level). Total: `O(n) × O(log n) = O(n log n)` across best, average, and worst cases.
- **Space — O(n):** Merging requires auxiliary arrays to hold the combined elements (plus O(log n) recursion stack).

**Comparison with Alternative Approaches**

| Approach | Best Time | Average Time | Worst Time | Extra Space | Stable? | Notes |
|---|---|---|---|---|---|---|
| **Bubble / Insertion Sort** | O(n) | O(n²) | O(n²) | O(1) | Yes | Easy to write, but too slow for `n > 10^4`. |
| **Quick Sort** | O(n log n) | O(n log n) | O(n²) | O(log n) | No | Fast in practice, but bad pivot causes O(n²). |
| **Merge Sort (Shown)** | **O(n log n)** | **O(n log n)** | **O(n log n)** | **O(n)** | **Yes** | Guaranteed O(n log n), ideal to write cleanly in interviews. |
| **Timsort (`sorted()`)** | **O(n)** | **O(n log n)** | **O(n log n)** | **O(n)** | **Yes** | Python's built-in; best choice when built-ins are allowed. |

## Rank 9 — Flatten a Nested List & Extract Strings

**Category:** Coding / Lists & Recursion · **Difficulty:** Easy · **Evidence [CR]:** Capgemini, HCLTech

**Problem Statement**
1. Flatten an arbitrarily deeply nested list/tuple into a single flat list.
2. Given `[[35, 66, 31], ["python", 13, "is"], [15, "fun", 14]]`, extract all strings and join them into `"python is fun"`.

**Examples**
- `list(flatten([1, [2, [3, 4]], (5,), 6]))` → `[1, 2, 3, 4, 5, 6]`
- String extraction → `"python is fun"`

```python
from collections.abc import Iterable, Iterator
from typing import Any

def flatten(items: Iterable[Any]) -> Iterator[Any]:
    """Lazily flatten arbitrarily nested lists and tuples."""
    for x in items:
        if isinstance(x, (list, tuple)):
            yield from flatten(x)
        else:
            yield x

def flatten_iterative(items: list[Any]) -> list[Any]:
    """Iterative stack-based flatten (safe for very deep nesting)."""
    result: list[Any] = []
    stack = list(reversed(items))
    while stack:
        curr = stack.pop()
        if isinstance(curr, (list, tuple)):
            stack.extend(reversed(curr))
        else:
            result.append(curr)
    return result

nested = [1, [2, [3, 4]], (5,), 6]
print("Recursive generator:", list(flatten(nested)))
print("Iterative stack:    ", flatten_iterative(nested))

my_list = [[35, 66, 31], ["python", 13, "is"], [15, "fun", 14]]
print("Extracted strings:  ", " ".join(x for sub in my_list for x in sub if isinstance(x, str)))
```

### Explanation & Complexity Analysis (Rank 9)

**How We Achieved This Solution**
- **Recursive Generator (`yield from`):** For each element `x`, if it is a `list` or `tuple`, we delegate recursion via `yield from flatten(x)`; otherwise we yield `x` directly. Importantly, we check `isinstance(x, (list, tuple))` rather than `isinstance(x, Iterable)` because strings are iterables—recursing on a string would infinitely recurse on 1-character strings!
- **Iterative Stack Alternative:** Pushing elements in reverse onto an explicit stack simulates recursion without hitting Python's ~1000-frame recursion limit.
- **1-Level Extraction:** When nesting depth is fixed at 2 levels, a nested generator expression inside `" ".join(...)` is cleaner and faster than recursion.

**How Time & Space Complexity Work**
- **Time — O(N + M):** Where `N` is the total number of leaf elements and `M` is the number of nested containers—every element and container is visited exactly once.
- **Space — O(d) extra (generator) / O(N) (materialized list):** The recursive generator only holds `d` call frames (where `d` is max nesting depth) at any time.

**Comparison with Alternative Approaches**

| Approach | Time Complexity | Extra Space | Handles Arbitrary Depth? | Notes |
|---|---|---|---|---|
| **Nested List Comprehension** | O(N) | O(N) | No (fixed depth) | Best for known 2D lists like `[[1, 2], [3, 4]]`. |
| **`sum(nested, [])` (Anti-pattern)** | O(N²) | O(N) | No (2D only) | Repeatedly copies lists—classic performance trap! |
| **Recursive Generator (`yield from`) (Optimal)** | **O(N + M)** | **O(d)** | **Yes** | Lazy, clean, Pythonic; watch recursion limit for `d > 1000`. |
| **Explicit Stack (Iterative)** | **O(N + M)** | **O(N)** | **Yes** | Avoids `RecursionError` on ultra-deep structures. |

## Rank 14 — Prime Numbers in a Range

**Category:** Coding / Loops & Math · **Difficulty:** Easy · **Evidence [CR]:** TCS

**Problem Statement**
1. Check whether a single integer `n` is prime.
2. Return all prime numbers in an inclusive range `[lo, hi]`.

**Examples**
- `is_prime(29)` → `True`
- `primes_in_range(10, 30)` → `[11, 13, 17, 19, 23, 29]`

```python
from math import isqrt

def is_prime(n: int) -> bool:
    """Trial division up to sqrt(n)."""
    if n < 2:
        return False
    if n in (2, 3):
        return True
    if n % 2 == 0 or n % 3 == 0:
        return False
    return all(n % d != 0 for d in range(5, isqrt(n) + 1, 2))

def primes_in_range(lo: int, hi: int) -> list[int]:
    """Sieve of Eratosthenes for inclusive range [lo, hi]."""
    if hi < 2:
        return []
    sieve = bytearray([1]) * (hi + 1)
    sieve[0:2] = b"\x00\x00"
    for p in range(2, isqrt(hi) + 1):
        if sieve[p]:
            # Mark all multiples of p starting at p*p as composite
            sieve[p * p :: p] = bytes(len(range(p * p, hi + 1, p)))
    return [i for i in range(max(lo, 2), hi + 1) if sieve[i]]

print("Is 29 prime?:", is_prime(29), "| Is 1 prime?:", is_prime(1))
print("Primes [10, 30]:", primes_in_range(10, 30))
```

### Explanation & Complexity Analysis (Rank 14)

**How We Achieved This Solution**
- **Single Number (`is_prime`):** If `n` has a factor larger than $\sqrt{n}$, its paired factor must be smaller than $\sqrt{n}$. Therefore we only test divisors up to `isqrt(n)`. Skipping even numbers cuts checks in half.
- **Range (`Sieve of Eratosthenes`):** Instead of testing each number independently, we create a compact `bytearray` marking every number from `0` to `hi` as potentially prime (`1`). For each prime `p <= sqrt(hi)`, we mark all multiples starting from `p * p` as composite (`0`) because smaller multiples (`2p, 3p, ..., (p-1)p`) were already crossed off by smaller primes. Slice assignment (`sieve[p*p::p] = ...`) executes at C speed.

**How Time & Space Complexity Work**
- **Trial Division (`is_prime`):** Tests at most $\sqrt{n}/2$ divisors → **O(√n) time, O(1) space**.
- **Sieve of Eratosthenes:** The total number of markings is $N \sum_{p \le \sqrt{N}}

**Comparison with Alternative Approaches**

| Approach (for Range `[1..N]`) | Time Complexity | Space Complexity | Trade-offs / Interview Notes |
|---|---|---|---|
| **Check `2..n-1` per number** | O(N²) | O(1) | Brute force; times out even for `N = 50,000`. |
| **Trial division `2..√n` per number** | O(N√N) | O(1) | Acceptable for small ranges (`N <= 10^4`) when memory is tight. |
| **Sieve of Eratosthenes (Optimal)** | **O(N log log N)** | **O(N)** | Near-linear time; standard optimization expected for range queries. |

## Rank 15 — Zip / Merge Two Lists & Dict from Two Lists

**Category:** Coding / Lists & Dicts · **Difficulty:** Easy · **Evidence [CR]:** TCS, Deloitte

**Problem Statement**
1. Pair elements from two lists (`zip`) and build a dictionary mapping keys to values.

- `dict(zip(["asha", "ravi"], [91, 78]))` → `{'asha': 91, 'ravi': 78}`
- `merge_sorted([1, 4, 9], [2, 3, 10])` → `[1, 2, 3, 4, 9, 10]`

```python
import heapq
from itertools import zip_longest

names = ["asha", "ravi", "meena"]
scores = [91, 78, 85]

pairs = list(zip(names, scores, strict=True))   # Raises ValueError if lengths differ (3.10+)
lookup = dict(zip(names, scores))
pass_list = [n for n, s in zip(names, scores) if s >= 80]
names_back, scores_back = zip(*pairs)           # Unzip

def merge_sorted_two_pointer(a: list[int], b: list[int]) -> list[int]:
    """Two-pointer O(n + m) merge of two sorted lists."""
    out: list[int] = []
    i = j = 0
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            out.append(a[i]); i += 1
        else:
            out.append(b[j]); j += 1
    out.extend(a[i:])
    out.extend(b[j:])
    return out

print("Lookup dict:", lookup)
print("Passed (>=80):", pass_list)
print("Unzipped:", names_back, scores_back)
print("Two-pointer merge:", merge_sorted_two_pointer([1, 4, 9], [2, 3, 10]))
print("heapq.merge:      ", list(heapq.merge([1, 4, 9], [2, 3, 10])))
print("zip_longest:      ", list(zip_longest([1, 2], ["a"], fillvalue="N/A")))
```

### Explanation & Complexity Analysis (Rank 15)

**How We Achieved This Solution**
- **`zip` Pitfall & Fix:** By default, `zip(a, b)` silently truncates to the shorter iterable—a common production bug. Mention `strict=True` (Python 3.10+) to catch mismatched lengths, or `itertools.zip_longest(..., fillvalue=...)` to pad shorter lists.
- **Unzipping:** `zip(*pairs)` unpacks the list of tuples as positional arguments, transposing rows into columns.
- **Merging Sorted Lists:** Using two pointers `i` and `j`, we compare `a[i]` and `b[j]` and append the smaller element, then append any leftover tail. `heapq.merge` generalizes this lazily to `k` sorted iterables.

**How Time & Space Complexity Work**
- **`zip` / `dict(zip(...))`:** `zip` itself is a lazy iterator (**O(1)** extra space). Materializing into a `list` or `dict` takes **O(min(n, m))** time and space.
- **Two-Pointer Sorted Merge:** Each element from `a` and `b` is appended once → **O(n + m) time, O(n + m) space** (output list).

**Comparison with Alternative Approaches (Merging Two Sorted Lists)**

| Approach | Time Complexity | Space Complexity | Trade-offs / Interview Notes |
|---|---|---|---|
| **`sorted(a + b)`** | O((n + m) log(n + m)) | O(n + m) | Ignores that inputs are already sorted (though Timsort partially optimizes this). |
| **Two-Pointer Merge (Optimal)** | **O(n + m)** | **O(n + m)** | Strict linear time; demonstrates algorithmic control. |

## Rank 17 — Two Sum

**Category:** Coding / Hashing · **Difficulty:** Easy · **Evidence [CR]:** HCLTech

**Problem Statement**
Given a list of integers `nums` and an integer `target`, return the indices `(i, j)` of the two distinct numbers that add up to `target` (or `None` if no pair exists).

**Example**
- `two_sum([2, 7, 11, 15], 9)` → `(0, 1)`

```python
def two_sum(nums: list[int], target: int) -> tuple[int, int] | None:
    """Single-pass hash map storing {value: index}."""
    seen: dict[int, int] = {}
    for i, x in enumerate(nums):
        complement = target - x
        if complement in seen:
            return seen[complement], i
        seen[x] = i
    return None

print(two_sum([2, 7, 11, 15], 9))
print(two_sum([3, 2, 4], 6))
print(two_sum([3, 3], 6))
```

### Explanation & Complexity Analysis (Rank 17)

**How We Achieved This Solution**
- For every element `x` at index `i`, we need another element equal to `complement = target - x`.
- Instead of scanning the rest of the array for `complement` (O(n) per element), we maintain a dictionary `seen` mapping each previously visited number to its index: `{value: index}`.
- **Order matters:** We check `if complement in seen` **before** inserting `seen[x] = i`. This guarantees an element can never pair with itself (e.g., `nums = [3, 2, 4], target = 6` won't wrongly return `(0, 0)`), while still handling duplicate values like `[3, 3], target = 6` cleanly.

**How Time & Space Complexity Work**
- **Time — O(n):** Single pass over `nums`; dictionary lookup (`in`) and insertion (`seen[x] = i`) are **O(1) average**.
- **Space — O(n):** In the worst case (pair is at the very end or doesn't exist), `seen` stores up to `n` entries.

**Comparison with Alternative Approaches**

| Approach | Time Complexity | Space Complexity | Trade-offs / Interview Notes |
|---|---|---|---|
| **Brute Force (Nested Loops)** | O(n²) | O(1) | Check every pair `(i, j)`. Mention this first, then optimize. |
| **Sort + Two Pointers** | O(n log n) | O(n) (to preserve indices) | Good if input is already sorted (Two Sum II) where space becomes O(1). |
| **One-Pass Hash Map (Optimal)** | **O(n)** | **O(n)** | Classic space-for-time trade-off; works on unsorted input with negatives/duplicates. |

## Rank 18 — Longest Substring Without Repeating Characters

**Category:** Coding / Sliding Window · **Difficulty:** Medium · **Evidence [CR]:** HCLTech

**Problem Statement**
Given a string `s`, return the length of the longest contiguous substring that contains no duplicate characters.

**Examples**
- `longest_unique_substring("abcabcbb")` → `3` (`"abc"`)
- `longest_unique_substring("pwwkew")` → `3` (`"wke"`)

```python
def longest_unique_substring(s: str) -> int:
    """Sliding window with last-seen index map."""
    last_seen: dict[str, int] = {}
    start = 0
    best = 0
    for i, ch in enumerate(s):
        # If ch was seen inside the current window [start, i), jump start past it
        if last_seen.get(ch, -1) >= start:
            start = last_seen[ch] + 1
        last_seen[ch] = i
        best = max(best, i - start + 1)
    return best

print("'abcabcbb' ->", longest_unique_substring("abcabcbb"))
print("'pwwkew'   ->", longest_unique_substring("pwwkew"))
print("'bbbbb'    ->", longest_unique_substring("bbbbb"))
print("''         ->", longest_unique_substring(""))
```

### Explanation & Complexity Analysis (Rank 18)

**How We Achieved This Solution**
- Maintain a valid sliding window `[start, i]` where all characters are unique, and a dictionary `last_seen` recording the most recent index of each character.
- As the right pointer `i` advances to `ch`:
  - If `ch` has appeared at or after `start` (`last_seen[ch] >= start`), we have a duplicate inside our window. Rather than shrinking `start` one character at a time, we **jump** `start` directly to `last_seen[ch] + 1`.
  - If `ch` appeared *before* `start`, it is outside the current window and can be safely ignored (`start` must never move backward!).
- Update `last_seen[ch] = i` and track `best = max(best, i - start + 1)`.

**How Time & Space Complexity Work**
- **Time — O(n):** The right pointer `i` moves from `0` to `n - 1` exactly once; each iteration does O(1) dictionary work.
- **Space — O(min(n, Σ)):** Where `Σ` is the character set size (e.g., 128 for ASCII), since `last_seen` holds at most one entry per unique character.

**Comparison with Alternative Approaches**

| Approach | Time Complexity | Space Complexity | Trade-offs / Interview Notes |
|---|---|---|---|
| **Brute Force (Check all substrings)** | O(n³) or O(n²) | O(min(n, Σ)) | Generates all O(n²) substrings and checks uniqueness. |
| **Sliding Window with `set`** | O(2n) = O(n) | O(min(n, Σ)) | Moves `start` step-by-step removing chars; each char visited at most twice. |
| **Sliding Window + Last-Seen Dict (Optimal)** | **O(n)** | **O(min(n, Σ))** | Jumps `start` in O(1); strictly single-pass over `s`. |

## Rank 19 — Remove Duplicates & Find Duplicates

**Category:** Coding / Lists & Sets · **Difficulty:** Easy · **Evidence [CR]:** HackerRank OA, Deloitte

**Problem Statement**
1. Remove duplicates from an unsorted list while preserving first-seen order.
2. Return all elements that appear more than once.
3. Remove duplicates **in place** from a *sorted* array using O(1) extra memory, returning the count `k` of unique elements.

**Examples**
- `dedupe([3, 1, 3, 2, 1])` → `[3, 1, 2]`
- `remove_dups_sorted_inplace([1, 1, 2, 3, 3, 3])` → `k = 3`, prefix `[1, 2, 3]`

```python
from collections import Counter
from typing import TypeVar

T = TypeVar("T")

def dedupe(xs: list[T]) -> list[T]:
    """Remove duplicates while preserving first-seen order."""
    return list(dict.fromkeys(xs))

def find_duplicates(xs: list[T]) -> list[T]:
    """Return elements appearing more than once."""
    return [x for x, count in Counter(xs).items() if count > 1]

def remove_dups_sorted_inplace(arr: list[int]) -> int:
    """Two-pointer in-place deduplication on a sorted array (O(1) extra space)."""
    if not arr:
        return 0
    write_idx = 1
    for read_idx in range(1, len(arr)):
        if arr[read_idx] != arr[write_idx - 1]:
            arr[write_idx] = arr[read_idx]
            write_idx += 1
    return write_idx

nums = [1, 1, 2, 3, 3, 3]
k = remove_dups_sorted_inplace(nums)
print("Dedupe (order kept):", dedupe([3, 1, 3, 2, 1]))
print("Duplicates found:   ", find_duplicates([1, 1, 2, 3, 3, 3]))
print(f"In-place sorted: k={k}, unique prefix={nums[:k]}")
```

### Explanation & Complexity Analysis (Rank 19)

**How We Achieved This Solution**
- **Order-Preserving Dedupe (`dict.fromkeys`):** Converting to `set(xs)` loses element order. Because Python 3.7+ dicts guarantee insertion order and keep unique keys, `list(dict.fromkeys(xs))` deduplicates in O(n) while keeping first-seen order.
- **In-Place Sorted Dedupe (Fast/Slow Two Pointers):** Since duplicates in a sorted array are adjacent, we keep a `write_idx` pointing to the next slot for a unique value and scan ahead with `read_idx`. Whenever `arr[read_idx] != arr[write_idx - 1]`, we've found a new unique value, so we write it at `arr[write_idx]` and advance `write_idx`.

**How Time & Space Complexity Work**
- **`dedupe` / `find_duplicates`:** Single pass with O(1) hash operations → **O(n) time, O(n) space**.
- **`remove_dups_sorted_inplace`:** Single pass modifying the input list in place → **O(n) time, O(1) extra space**.

**Comparison with Alternative Approaches**

| Approach | Time Complexity | Extra Space | Preserves Order? | Notes |
|---|---|---|---|---|
| **Loop with `if x not in res`** | O(n²) | O(n) | Yes | `x not in list` is O(n) per element—avoid! |
| **`list(set(xs))`** | O(n) | O(n) | **No** | Fast, but destroys original order. |
| **`list(dict.fromkeys(xs))` (Unsorted Optimal)** | **O(n)** | **O(n)** | **Yes** | Idiomatic, fast, order-preserving. |
| **Two Pointers (Sorted Optimal)** | **O(n)** | **O(1)** | **Yes** | Required when interviewer asks for O(1) space on sorted input. |

**How We Achieved This Solution**

- **Batch Queries (`string_anagram`):** Every anagram group shares the same **canonical signature**—either the sorted string (`"hack"` and `"khac"` both become `"achk"`) or a 26-element frequency tuple (for lowercase a–z). Precomputing a `Counter` of dictionary signatures lets us answer each query in O(1) lookup time (plus computing the query's signature).

**How Time & Space Complexity Work**
- **`is_anagram`:** Builds two frequency maps of length `n` → **O(n) time, O(k) space** (`k <= 26` for lowercase English).

## Rank 24 — Maximum Subarray Sum (Kadane's Algorithm)

**Category:** Coding / Arrays & DP · **Difficulty:** Medium · **Evidence [PP]:** HackerRank

**Problem Statement**
Given a non-empty list of integers (which may include negatives), find the contiguous subarray with the largest sum and return that sum.

**Examples**
- `max_subarray_sum([1, 2, 3, -2, 5])` → `9`
- `max_subarray_sum([-3, -1, -2])` → `-1`

```python
def max_subarray_sum(arr: list[int]) -> int:
    """Kadane's algorithm in O(n) time and O(1) space."""
    best = cur = arr[0]
    for x in arr[1:]:
        cur = max(x, cur + x)   # Either start fresh at x, or extend previous subarray
        best = max(best, cur)
    return best

print("[1, 2, 3, -2, 5] ->", max_subarray_sum([1, 2, 3, -2, 5]))
print("[-3, -1, -2]     ->", max_subarray_sum([-3, -1, -2]))
print("[-2,1,-3,4,-1,2,1,-5,4] ->", max_subarray_sum([-2, 1, -3, 4, -1, 2, 1, -5, 4]))
```

### Explanation & Complexity Analysis (Rank 24)

**How We Achieved This Solution**
- **Dynamic Programming Intuition:** Let `cur` be the maximum subarray sum *ending at the current index*. When we arrive at `x`, we have only two choices:
  1. Extend the previous subarray (`cur + x`), OR
  2. Abandon the previous subarray (because its sum is negative and would only drag us down) and start a new subarray at `x`.
- So `cur = max(x, cur + x)`, and `best = max(best, cur)` tracks the global maximum seen so far.
- **All-Negative Edge Case:** Initializing `best = cur = arr[0]` (instead of `0`) ensures that an all-negative array like `[-3, -1, -2]` returns `-1` rather than falsely returning `0`.

**How Time & Space Complexity Work**
- **Time — O(n):** Single left-to-right pass performing two `max()` comparisons per element.
- **Space — O(1):** Only two scalar variables (`cur` and `best`) are maintained instead of a full DP array.

**Comparison with Alternative Approaches**

| Approach | Time Complexity | Space Complexity | Trade-offs / Interview Notes |
|---|---|---|---|
| **Brute Force (3 nested loops)** | O(n³) | O(1) | Computes sum of every `(i, j)` pair from scratch. |
| **Prefix Sums (2 nested loops)** | O(n²) | O(1) or O(n) | Running sum per start index; still quadratic. |
| **Divide and Conquer** | O(n log n) | O(log n) | Splits into left, right, and crossing sums; more complex than needed. |
| **Kadane's Algorithm (Optimal)** | **O(n)** | **O(1)** | Optimal lower bound (must read every element at least once). |

## Rank 25 — Merge Intervals & Minimum CPU Cores

**Category:** Coding / Sorting & Sweep Line · **Difficulty:** Medium · **Evidence [PP + CR]:** HackerRank, Accenture

**Problem Statement**
1. Given a list of `[start, end]` intervals, merge all overlapping intervals.
2. Given process `start` and `end` times (inclusive), find the minimum number of CPU cores required so no two overlapping processes share a core.

**Examples**
- `merge_intervals([[1, 3], [2, 6], [8, 10], [15, 18]])` → `[[1, 6], [8, 10], [15, 18]]`
- `min_cores([1, 2, 5], [3, 4, 6])` → `2`

```python
def merge_intervals(intervals: list[list[int]]) -> list[list[int]]:
    """Sort by start time and greedily merge overlapping intervals."""
    merged: list[list[int]] = []
    for start, end in sorted(intervals):
        if merged and start <= merged[-1][1]:
            merged[-1][1] = max(merged[-1][1], end)
        else:
            merged.append([start, end])
    return merged

def min_cores(start: list[int], end: list[int]) -> int:
    """Sweep-line event counting; (time, 0)=start sorts before (time, 1)=end for inclusive overlap."""
    events = sorted([(s, 0) for s in start] + [(e, 1) for e in end])
    cur = best = 0
    for _, kind in events:
        if kind == 0:
            cur += 1
            best = max(best, cur)
        else:
            cur -= 1
    return best

print("Merged:", merge_intervals([[1, 3], [2, 6], [8, 10], [15, 18]]))
print("Min cores:", min_cores([1, 2, 5], [3, 4, 6]))
print("Inclusive touch [1,3] & [3,5]:", min_cores([1, 3], [3, 5]))
```

### Explanation & Complexity Analysis (Rank 25)

**How We Achieved This Solution**
- **Merge Intervals:** Sorting intervals by start time ensures any intervals that overlap appear consecutively. We only need to compare the current interval's `start` with the last merged interval's `end` (`merged[-1][1]`). If `start <= merged[-1][1]`, extend the end with `max(merged[-1][1], end)`.
- **Minimum CPU Cores (Sweep Line):** Every process adds `+1` active core at its start and `-1` at its end. Sorting all `2n` events chronologically and tracking the peak active count gives the exact minimum cores needed.
- **Tie-Breaking Rule:** Using `(s, 0)` for starts and `(e, 1)` for ends forces starts at timestamp `t` to be processed *before* ends at timestamp `t`, which correctly models **inclusive** endpoints (a process ending at `t` still overlaps one starting at `t`). If endpoints are half-open `[start, end)`, swap `0` and `1` so ends free a core first.

**How Time & Space Complexity Work**
- **Time — O(n log n):** Dominated by sorting `n` intervals (or `2n` events); the linear scan afterward is O(n).
- **Space — O(n):** Required to store the sorted output/events list.

**Comparison with Alternative Approaches**

| Approach | Time Complexity | Space Complexity | Trade-offs / Interview Notes |
|---|---|---|---|
| **Pairwise Overlap Check** | O(n²) | O(n) | Compares every interval against all others; too slow. |
| **Min-Heap of End Times (`min_cores`)** | O(n log n) | O(n) | Sort by start, push end times into a min-heap; pop finished cores. Same Big-O as sweep line. |
| **Sort + Linear Scan / Sweep Line (Optimal)** | **O(n log n)** | **O(n)** | Shortest, cleanest code; tie-breaking is explicit and easy to explain. |

## Rank 26 — Product of Array Except Self

**Category:** Coding / Prefix-Suffix · **Difficulty:** Medium · **Evidence [CR]:** EY

**Problem Statement**
Given an integer array `nums`, return an array `out` where `out[i]` is the product of all elements of `nums` except `nums[i]`, **without using division** and in **O(n)** time.

**Example**
- `product_except_self([1, 2, 3, 4])` → `[24, 12, 8, 6]`
- `product_except_self([-1, 1, 0, -3, 3])` → `[0, 0, 9, 0, 0]`

```python
def product_except_self(nums: list[int]) -> list[int]:
    """Two-pass prefix and suffix product in O(n) time, O(1) extra space."""
    n = len(nums)
    out = [1] * n

    # Pass 1: out[i] stores product of all elements strictly to the left of i
    prefix = 1
    for i in range(n):
        out[i] = prefix
        prefix *= nums[i]

    # Pass 2: multiply out[i] by product of all elements strictly to the right of i
    suffix = 1
    for i in range(n - 1, -1, -1):
        out[i] *= suffix
        suffix *= nums[i]

    return out

print("[1, 2, 3, 4]       ->", product_except_self([1, 2, 3, 4]))
print("[-1, 1, 0, -3, 3]  ->", product_except_self([-1, 1, 0, -3, 3]))
```

### Explanation & Complexity Analysis (Rank 26)

**How We Achieved This Solution**
- For any index `i`, the product of all elements except `nums[i]` equals:
  $$(	ext{product of everything to the left of } i) 	imes (	ext{product of everything to the right of } i)$$
- **Pass 1 (Left → Right):** Maintain a running `prefix` product. Store `prefix` into `out[i]` *before* multiplying `prefix *= nums[i]`.
- **Pass 2 (Right → Left):** Maintain a running `suffix` product. Multiply `out[i] *= suffix` *before* updating `suffix *= nums[i]`.
- Because no division is used, arrays containing one or multiple zeros work naturally without special-case `ZeroDivisionError` guards.

**How Time & Space Complexity Work**
- **Time — O(n):** Two linear passes (`2n` multiplications).
- **Space — O(1) extra space:** Excluding the required output array `out`, we only use two scalar variables (`prefix` and `suffix`).

**Comparison with Alternative Approaches**

| Approach | Time Complexity | Extra Space | Handles Zeros? | Notes |
|---|---|---|---|---|
| **Brute Force (Nested Loops)** | O(n²) | O(1) | Yes | Multiplies `n - 1` elements for every index. |
| **Total Product ÷ `nums[i]`** | O(n) | O(1) | Needs special cases | Breaks on zeros and violates the "no division" constraint. |
| **Separate `left[]` & `right[]` Arrays** | O(n) | O(n) | Yes | Easy to understand first, then optimize by folding into `out`. |
| **In-Place Prefix/Suffix on `out` (Optimal)** | **O(n)** | **O(1)** | **Yes** | Meets both O(n) time and O(1) extra-space follow-up constraints. |

```python
import math
from dataclasses import dataclass

class Rectangle:
    def __init__(self, length: float, width: float) -> None:
        self.length = length
        self.width = width

    def area(self) -> float:
        return self.length * self.width

class Circle:
    def __init__(self, radius: float) -> None:
        self.radius = radius

    def area(self) -> float:
        return math.pi * (self.radius ** 2)

@dataclass(frozen=True, slots=True)
class Item:
    name: str
    price: int

class ShoppingCart:
    def __init__(self) -> None:
        self._items: list[Item] = []
        self._running_total: int = 0

    def add(self, item: Item) -> None:
        self._items.append(item)
        self._running_total += item.price

    def total(self) -> int:
        return self._running_total

    def __len__(self) -> int:
        return len(self._items)

print(f"Rectangle: {Rectangle(2, 3).area():.2f} | Circle: {Circle(1).area():.2f}")
cart = ShoppingCart()
cart.add(Item("Bike", 1000))
cart.add(Item("Lock", 50))
print("Cart len:", len(cart), "| Total:", cart.total())
```

### Explanation & Complexity Analysis (Rank 28)

**How We Achieved This Solution**
- **Dunder Methods (`__len__`):** HackerRank's locked test harness calls `len(cart)` directly. Implementing `__len__` plugs your class into Python's data model.
- **Maintaining a Running Total:** Instead of recomputing `sum(i.price for i in self._items)` on every `total()` call (O(n)), maintaining `self._running_total` on each `add()` makes `total()` **O(1)**!
- **`@dataclass(frozen=True, slots=True)`:** Makes `Item` immutable, hashable, concise, and memory-efficient.

**How Time & Space Complexity Work**
- **`Rectangle.area()` / `Circle.area()`:** **O(1) time, O(1) space**.
- **`ShoppingCart`:** `add()` → **O(1) amortized**, `total()` → **O(1)** (with running total), `__len__()` → **O(1)**. Space is **O(n)** for `n` items stored.

**Comparison with Alternative Approaches**

| Approach for `ShoppingCart.total()` | `add()` Time | `total()` Time | Trade-offs / Interview Notes |
|---|---|---|---|
| **Recompute `sum(...)` each call** | O(1) | O(n) | Fine for small carts, slow if `total()` is queried repeatedly in a loop. |

## Rank 29 — Multiset Implementation (OOP + Dunder Methods)

**Category:** Coding / OOP & Dunder Methods · **Difficulty:** Easy · **Evidence [PP]:** HackerRank

**Problem Statement**
Implement a `Multiset` class supporting:
- `add(val)`: adds one occurrence of `val`.
- `remove(val)`: removes one occurrence of `val` if present (no-op if absent).
- `val in m` (`__contains__`): returns `True` if at least one occurrence exists.
- `len(m)` (`__len__`): returns total count of elements (including duplicates).

**Example**
- `add(1), add(1), remove(1)` → `1 in m` is `True`, `len(m)` is `1`.

```python
from collections import Counter

class Multiset:
    def __init__(self) -> None:
        self._counts: Counter[int] = Counter()
        self._size: int = 0

    def add(self, val: int) -> None:
        self._counts[val] += 1
        self._size += 1

    def remove(self, val: int) -> None:
        if self._counts[val] > 0:
            self._counts[val] -= 1
            if self._counts[val] == 0:
                del self._counts[val]  # Prevents stale 0-count keys from leaking memory
            self._size -= 1

    def __contains__(self, val: int) -> bool:
        return self._counts[val] > 0

    def __len__(self) -> int:
        return self._size

m = Multiset()
m.add(1); m.add(1); m.remove(1)
print("1 in m:", 1 in m, "| len(m):", len(m))
m.remove(1); m.remove(99)
print("1 in m after 2nd remove:", 1 in m, "| len(m):", len(m))
```

### Explanation & Complexity Analysis (Rank 29)

**How We Achieved This Solution**
- Rather than storing every duplicate item in a `list`, we store a frequency map (`Counter`) plus an integer `_size` counter.
- Accessing a missing key in `Counter` returns `0` without inserting it. Deleting keys when their count drops to `0` keeps memory strictly proportional to currently active distinct values.
- Tracking `_size` manually avoids summing `self._counts.values()` inside `__len__`, keeping `len(m)` O(1).

**How Time & Space Complexity Work**
- **Time:** `add`, `remove`, `__contains__`, and `__len__` are all **O(1) average**.
- **Space — O(k):** Where `k` is the number of *distinct* active values (can be much smaller than total elements `n` if there are many duplicates).

**Comparison with Alternative Approaches**

| Approach | `add` | `remove` | `in` (`__contains__`) | `len` | Space | Notes |
|---|---|---|---|---|---|---|
| **Backed by `list`** | O(1) | O(n) | O(n) | O(1) | O(n) | Common naive solution; linear scans degrade under heavy queries. |
| **Backed by `Counter` + `_size` (Optimal)** | **O(1)** | **O(1)** | **O(1)** | **O(1)** | **O(k)** | Strong 3-year signal: turns O(n) operations into O(1). |

## Rank 30 — Dominant Cells (2D Matrix)

**Category:** Coding / 2D Arrays · **Difficulty:** Easy · **Evidence [PP]:** HackerRank

**Problem Statement**
Given an `R × C` grid of integers, count how many cells are **strictly greater** than all of their existing neighbors (up to 8 neighbors: horizontal, vertical, and diagonal).

**Example**
- `[[1, 2, 7], [4, 5, 6], [8, 8, 9]]` → `2` (cells `7` and `9`; both `8`s fail because they tie each other and neighbor `9`).

```python
def num_cells(grid: list[list[int]]) -> int:
    rows, cols = len(grid), len(grid[0])

    def is_dominant(r: int, c: int) -> bool:
        val = grid[r][c]
        for dr in (-1, 0, 1):
            for dc in (-1, 0, 1):
                if dr == 0 and dc == 0:
                    continue
                nr, nc = r + dr, c + dc
                if 0 <= nr < rows and 0 <= nc < cols and grid[nr][nc] >= val:
                    return False
        return True

    return sum(is_dominant(r, c) for r in range(rows) for c in range(cols))

print("Dominant cells:", num_cells([[1, 2, 7], [4, 5, 6], [8, 8, 9]]))
```

### Explanation & Complexity Analysis (Rank 30)

**How We Achieved This Solution**
- Using direction offsets `dr, dc ∈ {-1, 0, 1}` (skipping `(0, 0)`) cleanly generates all 8 surrounding coordinates without writing 8 repetitive `if` statements.
- Bounds checks `0 <= nr < rows and 0 <= nc < cols` handle corners (3 neighbors) and edges (5 neighbors) uniformly.
- Early exit (`return False` on the first neighbor `>= val`) avoids checking remaining neighbors once a cell is disqualified. Note `>=` enforces **strictly greater** (ties fail).

**How Time & Space Complexity Work**
- **Time — O(R · C):** Each of the `R · C` cells checks at most 8 constant-time neighbors → `8 · R · C = O(R · C)`.
- **Space — O(1) extra space:** Only loop counters and scalar values are used.

**Comparison with Alternative Approaches**

| Approach | Time Complexity | Extra Space | Trade-offs / Interview Notes |
|---|---|---|---|
| **Padding Grid with `-inf` Border** | O(R · C) | O(R · C) | Avoids bounds checks, but allocates a full copy of the grid. |
| **Direct 8-Direction Offset Check (Optimal)** | **O(R · C)** | **O(1)** | Optimal in both time and space; short-circuits immediately on failure. |

## Rank 42 — Balanced Parentheses

**Category:** Coding / Stack · **Difficulty:** Easy · **Evidence [CR]:** Infosys, GeeksforGeeks

**Problem Statement**
Given a string containing brackets `(`, `)`, `[`, `]`, `{`, `}`, determine if every opening bracket is closed by the same type of bracket in the correct LIFO order.

**Examples**
- `is_balanced("{[()]}")` → `True`
- `is_balanced("([)]")` → `False`

```python
def is_balanced(s: str) -> bool:
    pairs = {")": "(", "]": "[", "}": "{"}
    stack: list[str] = []
    for ch in s:
        if ch in "([{":
            stack.append(ch)
        elif ch in pairs:
            if not stack or stack.pop() != pairs[ch]:
                return False
    return not stack

for test in ["{[()]}", "([)]", "(((", "())", ""]:
    print(f"{test!r:10} -> {is_balanced(test)}")
```

### Explanation & Complexity Analysis (Rank 42)

**How We Achieved This Solution**
- The most recently opened bracket must be the first one closed—a textbook **LIFO (Last-In, First-Out)** stack problem.
- Push every opener onto `stack`. When encountering a closer, look up its expected opener in `pairs`. If the stack is empty (stray closer like `")"`) or the popped opener doesn't match, return `False` immediately.
- At the end, `return not stack` catches unclosed openers (like `"((("`).

**How Time & Space Complexity Work**
- **Time — O(n):** Each character is pushed and popped at most once; `list.append` and `list.pop()` at the end of a Python list are O(1) amortized.
- **Space — O(n):** Worst case (all opening brackets like `"(((((("`) stores all `n` characters on the stack.

**Comparison with Alternative Approaches**

| Approach | Time Complexity | Space Complexity | Trade-offs / Interview Notes |
|---|---|---|---|
| **Repeated `replace("()", "")` loop** | O(n²) | O(n) | Creates new strings on every pass; fails on large inputs. |
| **Single Counter (`+1`/`-1`)** | O(n) | O(1) | Works **only** for a single bracket type `()`; fails on `"([)]"`. |
| **Stack + Matching Dict (Optimal)** | **O(n)** | **O(n)** | Handles arbitrary nested bracket types in a single pass. |

## Rank 43 — GCD of an Array & Decimal to Binary

**Category:** Coding / Math · **Difficulty:** Easy · **Evidence [CR]:** Cognizant, LTIMindtree

**Problem Statement**
1. Compute the Greatest Common Divisor (GCD) of all integers in an array.
2. Convert a non-negative integer from decimal to its binary string representation without relying solely on `bin()`.

**Examples**
- `gcd_array([12, 18, 24])` → `6`
- `to_binary(10)` → `"1010"`

```python
import math
from functools import reduce

def gcd_two(a: int, b: int) -> int:
    """Euclidean algorithm by hand."""
    while b:
        a, b = b, a % b
    return abs(a)

def gcd_array(arr: list[int]) -> int:
    return reduce(gcd_two, arr)      # Built-in shortcut on 3.9+: math.gcd(*arr)

def to_binary(n: int) -> str:
    if n == 0:
        return "0"
    bits: list[str] = []
    while n > 0:
        bits.append(str(n % 2))
        n //= 2
    return "".join(reversed(bits))

print("GCD [12, 18, 24]:", gcd_array([12, 18, 24]), "| Built-in:", math.gcd(12, 18, 24))
print("Binary of 10:", to_binary(10), "| Built-in:", format(10, "b"))
print("Binary of 0: ", to_binary(0))
```

### Explanation & Complexity Analysis (Rank 43)

**How We Achieved This Solution**
- **Euclidean Algorithm (`gcd(a, b) = gcd(b, a % b)`):** Because any common divisor of `a` and `b` also divides `a % b`, repeatedly replacing `(a, b)` with `(b, a % b)` shrinks the numbers rapidly until `b == 0`. Applying `reduce` folds this associatively across the entire array (and can short-circuit if the running GCD ever hits `1`).
- **Decimal to Binary:** Repeatedly taking `n % 2` extracts bits from least significant to most significant, while `n //= 2` shifts right. Collecting bits in a list and reversing once avoids O(k²) string prepending.

**How Time & Space Complexity Work**
- **GCD of Array:** Each pairwise GCD takes O(log M) steps (where `M` is the max element). Over `n` elements: **O(n + log M)** amortized (bounded by **O(n log M)**), **O(1) space**.
- **Decimal to Binary:** `n` halves each iteration → **O(log n) time, O(log n) space** (since `n` has $\lfloor \log_2 n 
floor + 1$ bits).

**Comparison with Alternative Approaches**

| Problem / Approach | Time Complexity | Space Complexity | Trade-offs / Interview Notes |
|---|---|---|---|
| **GCD: Trial division `min(arr)` down to `1`** | O(n · M) | O(1) | Exponentially slower than Euclid for large values. |
| **GCD: Euclidean Algorithm (Optimal)** | **O(n log M)** | **O(1)** | Standard optimal method; mention `math.gcd(*arr)` in Python 3.9+. |
| **Binary: String prepend `res = bit + res`** | O((log n)²) | O(log n) | Copies string on each bit; list + reverse is cleaner. |
| **Binary: List + Reverse / `format(n, 'b')` (Optimal)** | **O(log n)** | **O(log n)** | Show manual loop first, then mention `bin(n)[2:]` / `format(n, 'b')`. |

## Rank 44 — Desired Array (Sum of `k` Smallest Non-Divisible Positives)

**Category:** Coding / Math & Loops · **Difficulty:** Medium · **Evidence [CR]:** Accenture

**Problem Statement**
Given an array `arr` of positive integers and an integer `k`, find the sum of the `k` smallest positive integers that are **not** divisible by any element in `arr`.

**Example**
- `desired_sum([2, 3, 4, 5, 6], 4)` → `1 + 7 + 11 + 13 = 32`

```python
def desired_sum(arr: list[int], k: int) -> int:
    """Prune redundant divisors, then scan positive integers."""
    raw = sorted(set(arr))
    if not raw or raw[0] == 1:
        return 0  # Every positive integer is divisible by 1

    # Optimization: drop any divisor divisible by a smaller divisor (e.g., 4 & 6 redundant with 2)
    divisors: list[int] = []
    for d in raw:
        if all(d % prev != 0 for prev in divisors):
            divisors.append(d)

    total = found = n = 0
    while found < k:
        n += 1
        if all(n % d != 0 for d in divisors):
            total += n
            found += 1
    return total

print("k=4, arr=[2,3,4,5,6] ->", desired_sum([2, 3, 4, 5, 6], 4))
print("Edge case with 1     ->", desired_sum([1, 2, 3], 4))
```

### Explanation & Complexity Analysis (Rank 44)

**How We Achieved This Solution**
- **Divisor Pruning:** If `2` is in `arr`, then `4`, `6`, `8` can never eliminate a number that `2` hasn't already eliminated. Sorting unique divisors and keeping only minimal (primitive) divisors shrinks the inner check dramatically (`[2, 3, 4, 5, 6]` reduces to `[2, 3, 5]`).
- **Guard Against `1`:** Every integer is divisible by `1`, so if `1` is present, no valid numbers exist—returning `0` immediately prevents an infinite loop.
- **Candidate Scan:** Increment `n` from `1` upward; whenever `all(n % d != 0 for d in divisors)` holds, add `n` to `total` until `found == k`.

**How Time & Space Complexity Work**
- **Time — O(N_k · |D_min|):** Where `N_k` is the `k`-th valid integer and `|D_min|` is the count of pruned divisors (plus O(m log m) to sort `arr`).
- **Space — O(m):** To store the unique/pruned divisors.

**Comparison with Alternative Approaches**

| Approach | Time Complexity | Space Complexity | Trade-offs / Interview Notes |
|---|---|---|---|
| **Unpruned Brute Force** | O(N_k · |arr|) | O(1) | Checks duplicates and redundant multiples on every candidate. |
| **Pruned Divisors + Scan (Shown)** | **O(N_k · |D_min|)** | **O(|arr|)** | Clean, fast, and easily passes typical OA constraints. |
| **Binary Search + Inclusion-Exclusion** | O(2^|D_min| · log(N_k)) | O(|D_min|) | Required only if `k` is huge (`k = 10^9`) and `|D_min|` is small (`<= 20`). |

## Rank 45 — Count Subsequences of Equal Elements

**Category:** Coding / Combinatorics & Hashing · **Difficulty:** Medium · **Evidence [CR]:** Deloitte

**Problem Statement**
Given an array of integers `arr`, return the total number of non-empty subsequences in which all elements are equal, modulo $10^9 + 7$.

**Example**
- `count_equal_subsequences([1, 2, 1])` → `4` (`{1}`, `{1}`, `{1, 1}`, `{2}`)

```python
from collections import Counter

def count_equal_subsequences(arr: list[int]) -> int:
    MOD = 10**9 + 7
    counts = Counter(arr)
    return sum((pow(2, c, MOD) - 1) for c in counts.values()) % MOD

print("[1, 2, 1]       ->", count_equal_subsequences([1, 2, 1]))
print("[5, 5, 5]       ->", count_equal_subsequences([5, 5, 5]))
print("[1, 2, 3, 4]    ->", count_equal_subsequences([1, 2, 3, 4]))
```

### Explanation & Complexity Analysis (Rank 45)

**How We Achieved This Solution**
- A subsequence with all equal elements can only be formed by choosing a non-empty subset of occurrences of a single distinct value.
- If a value appears `c` times in `arr`, each occurrence can be either included or excluded, giving $2^c$ subsets. Subtracting `1` removes the empty subset, leaving $2^c - 1$ valid subsequences for that value.
- Summing $(2^c - 1) mod (10^9 + 7)$ over all distinct values gives the exact answer. Using Python's 3-argument `pow(2, c, MOD)` performs modular exponentiation in O(log c) without constructing huge integers like `2 ** 100000`.

**How Time & Space Complexity Work**
- **Time — O(n + k log c):** `Counter(arr)` takes O(n); computing `pow(2, c, MOD)` for each of `k` distinct values takes O(log c) each (total strictly O(n)).
- **Space — O(k):** Stores frequency counts for `k` distinct elements.

**Comparison with Alternative Approaches**

| Approach | Time Complexity | Space Complexity | Trade-offs / Interview Notes |
|---|---|---|---|
| **Generate All Subsequences** | O(2^n) | O(n) | Exponential—completely infeasible for `n > 25`. |
| **`(2 ** c - 1) % MOD` (without 3-arg `pow`)** | O(n²) worst (big-int) | O(n) | `2 ** 100000` allocates a massive integer before taking `% MOD`. |
| **`Counter` + `pow(2, c, MOD)` (Optimal)** | **O(n)** | **O(k)** | Linear time, constant-sized intermediate ints via modular exponentiation. |

## Rank 2 — Decorators (Write One + Parameterized Decorator)

**Category:** Theory + Code / Functions · **Difficulty:** Medium · **Evidence [CR]:** Accenture, Wipro, Infosys, Capgemini

```python
import functools
import time
from collections.abc import Callable
from typing import Any

def timed(func: Callable[..., Any]) -> Callable[..., Any]:
    @functools.wraps(func)   # Preserves __name__, __doc__, and metadata
    def wrapper(*args: Any, **kwargs: Any) -> Any:
        start = time.perf_counter()
        try:
            return func(*args, **kwargs)
        finally:
            elapsed = time.perf_counter() - start
            print(f"{func.__name__} took {elapsed:.5f}s")
    return wrapper

def retry(times: int) -> Callable[..., Any]:
    """Parameterized decorator (3 levels of nesting)."""
    def decorator(func: Callable[..., Any]) -> Callable[..., Any]:
        @functools.wraps(func)
        def wrapper(*args: Any, **kwargs: Any) -> Any:
            for attempt in range(1, times + 1):
                try:
                    return func(*args, **kwargs)
                except Exception:
                    if attempt == times:
                        raise
        return wrapper
    return decorator

@timed
def fast_add(a: int, b: int) -> int:
    return a + b

print("Result:", fast_add(10, 20), "| Name preserved:", fast_add.__name__)
```

**Explanation:** A decorator is a callable that takes a function and returns a wrapped replacement (`@timed` is syntactic sugar for `fast_add = timed(fast_add)`). Always include `@functools.wraps(func)` so debugging tools and docstrings still see the original function name. Common production uses: logging, timing, auth (`@login_required`), routing (`@app.route`), and caching (`@functools.cache`).

## Rank 5 — OOP: `__init__`, `self`, Multiple Inheritance & MRO (Diamond Problem)

**Category:** Theory + Code / OOP · **Difficulty:** Medium · **Evidence [CR]:** Infosys, Wipro, EY

```python
class A:
    def hello(self) -> str:
        return "A"

class B(A):
    def hello(self) -> str:
        return "B" + super().hello()

class C(A):
    def hello(self) -> str:
        return "C" + super().hello()

class D(B, C):
    pass

print("D().hello():", D().hello())
print("MRO:", [cls.__name__ for cls in D.__mro__])
```

**Explanation:**
- **`__new__` vs `__init__`:** `__new__` allocates and returns the new instance; `__init__` initializes its attributes afterward.
- **`self`:** Explicit reference to the instance passed as the first argument to instance methods.
- **MRO & Diamond Inheritance:** Python resolves method lookup using **C3 Linearization** (`D.__mro__` = `[D, B, C, A, object]`). Crucially, `super()` means *"next class in the MRO"*, not *"my direct parent"*—so `B.hello()` calls `C.hello()`, which calls `A.hello()` **once**, yielding `"BCA"`.
- **Method Overloading:** Python does not overload by signature; use default parameters, `*args`, or `@functools.singledispatchmethod`.

## Rank 6 — Generators, Iterators & Comprehensions

**Category:** Theory + Code / Iteration · **Difficulty:** Medium · **Evidence [CR]:** Capgemini, Infosys, LTIMindtree

```python
from collections.abc import Iterator

def fib_generator(limit: int) -> Iterator[int]:
    """Generator function yielding values lazily in O(1) memory."""
    a, b = 0, 1
    for _ in range(limit):
        yield a
        a, b = b, a + b

class Countdown:
    """Custom iterator implementing the iterator protocol manually."""
    def __init__(self, start: int) -> None:
        self.current = start

    def __iter__(self) -> "Countdown":
        return self

    def __next__(self) -> int:
        if self.current <= 0:
            raise StopIteration
        val = self.current
        self.current -= 1
        return val

print("Generator:", list(fib_generator(6)))
print("Countdown:", list(Countdown(3)))
```

**Explanation:**
- **Iterable vs Iterator:** An **iterable** implements `__iter__()` (returns an iterator); an **iterator** implements both `__iter__()` and `__next__()` (raising `StopIteration` when exhausted).
- **Why Generators Matter:** A function with `yield` pauses its stack frame between calls and produces items on demand in **O(1) memory**—the textbook answer for *"how do you process a 50 GB log file without running out of RAM?"* (`yield from` delegates to a sub-generator).

## Rank 21 — Memory Management, GIL & Multithreading vs Multiprocessing

**Category:** Theory / Internals & Concurrency · **Difficulty:** Medium · **Evidence [CR]:** TCS, Infosys, Deloitte

- **Memory Management (CPython):**
  - Objects live on a private heap managed by Python's memory manager.
  - Primary reclamation is **reference counting** (objects are freed immediately when their refcount hits `0`).
  - A **generational cyclic garbage collector** (`gc` module) detects and collects reference cycles (e.g., `a.ref = b; b.ref = a`).
- **Global Interpreter Lock (GIL):**
  - Ensures only one thread executes Python bytecode at a time inside a CPython process, protecting internal C-level object reference counts.
  - **I/O-bound tasks** (HTTP calls, DB queries, file I/O): `threading` or `asyncio` works great because threads release the GIL while waiting on I/O.
  - **CPU-bound tasks** (number crunching, image/ML loops): use `multiprocessing` / `ProcessPoolExecutor` (separate processes, separate GILs) or C-extension libraries like NumPy.
  - **Modern Bonus (Python 3.13 / 3.14):** Free-threaded CPython (no-GIL build) was introduced experimentally in 3.13 (PEP 703) and became officially supported (optional build) in 3.14 (PEP 779).

## Rank 41 — Debugging: HackerRank "Default Arguments" (`print_from_stream`)

**Category:** Debugging / Mutability · **Difficulty:** Medium · **Evidence [PP]:** HackerRank

```python
class EvenStream:
    def __init__(self) -> None:
        self.current = 0
    def get_next(self) -> int:
        val = self.current
        self.current += 2
        return val

class OddStream:
    def __init__(self) -> None:
        self.current = 1
    def get_next(self) -> int:
        val = self.current
        self.current += 2
        return val

# Buggy signature: def print_from_stream(n, stream=EvenStream()):
# Fix: default to None and instantiate a fresh EvenStream() on each call
def print_from_stream(n: int, stream: EvenStream | OddStream | None = None) -> list[int]:
    if stream is None:
        stream = EvenStream()
    return [stream.get_next() for _ in range(n)]

print("Call 1 (default even, 3):", print_from_stream(3))
print("Call 2 (default even, 3):", print_from_stream(3))  # Freshly restarts at 0!
print("Call 3 (odd, 3):         ", print_from_stream(3, OddStream()))
```

**Explanation:** This is the mutable-default-argument trap (Rank 10) applied to a stateful object. When `stream=EvenStream()` is in the signature, one `EvenStream` instance is created once at function definition time and shared across all calls that omit `stream`—so the second call continues from `6` instead of restarting at `0`. Using `stream=None` creates a fresh instance per call.

## Real Assessment Q1 — Lexicographical Binary Substring Replacement

**Category:** Strings / Single-Pass Simulation · **Difficulty:** Easy–Medium · **Evidence [CR]:** Accenture (AI Engineer OA)

**Problem Statement**
Given a binary string `s` of length `n` (`1 <= n <= 10^5`), process `s` from left to right starting at `i = 0`. While `i <= n - 3`, inspect `s[i:i+3]`:
1. If `"110"` → emit `"101"` and advance `i += 3`.
2. If `"111"` → emit `"111"` unchanged and advance `i += 3`.
3. Otherwise → emit `s[i]` unchanged and advance `i += 1`.

When `i > n - 3`, append any remaining tail characters unchanged and return the transformed string.

**Trace Example (`"011001111100"` → `"010101111010"`)**

| `i` | `s[i:i+3]` | Action | Next `i` |
|---|---|---|---|
| 0 | `011` | no match → keep `'0'` | 1 |
| 1 | `110` | emit `'101'` | 4 |
| 4 | `011` | no match → keep `'0'` | 5 |
| 5 | `111` | emit `'111'` | 8 |
| 8 | `110` | emit `'101'` | 11 |
| 11 | (`> n-3`) | loop ends, append tail `'0'` | — |

```python
def normalize_binary(s: str) -> str:
    """Single-pass left-to-right simulation in O(n) time and O(n) space."""
    n = len(s)
    out: list[str] = []
    i = 0
    while i <= n - 3:
        chunk = s[i : i + 3]
        if chunk == "110":
            out.append("101")
            i += 3
        elif chunk == "111":
            out.append("111")
            i += 3
        else:
            out.append(s[i])
            i += 1
    out.append(s[i:])  # Append remaining 0..2 tail characters
    return "".join(out)

tests = [
    ("110", "101"),
    ("110111", "101111"),
    ("011001111100", "010101111010"),
    ("1110", "1110"),       # Trap: '111' consumes first 3 chars, so '110' is skipped!
    ("11110", "11110"),     # Trap: same skip rule
    ("11", "11"),           # n < 3 edge case
]
for inp, expected in tests:
    got = normalize_binary(inp)
    assert got == expected, f"Mismatch for {inp}: {got} != {expected}"
    print(f"{inp:14} -> {got} (OK)")
```

### Explanation & Complexity Analysis (Real Assessment Q1)

**How We Achieved This Solution**
- The problem is a **deterministic finite cursor simulation**, NOT a global search-and-replace.
- Using an explicit `while i <= n - 3` pointer lets us advance by `+3` after matching `"110"` or `"111"`, and by `+1` otherwise.
- Collecting output chunks in a Python `list` and joining once at the end avoids quadratic string copying on `n = 10^5`.

**Critical Traps**
- **`s.replace("110", "101")` is WRONG:** On `"1110"`, the spec matches `"111"` at index `0` and jumps to index `3`, leaving `"1110"` unchanged. `s.replace()` ignores the `"111"` skip rule and wrongly returns `"1101"`.
- **Don't re-scan replaced characters:** Advancing by `3` ensures `"101"` is never re-inspected.
- **Tail handling:** Always append `s[i:]` after the loop so the final 0–2 characters (or strings with `n < 3`) aren't dropped.

**How Time & Space Complexity Work**
- **Time — O(n):** `i` strictly increases by at least `1` every iteration, doing O(1) slice (length 3) and list append work → total O(n), instant even at `n = 10^5`.
- **Space — O(n):** `out` stores chunks totaling `n` characters, plus the final joined string of length `n`.

**Comparison with Alternative Approaches**

| Approach | Correct? | Time Complexity | Space Complexity | Notes |
|---|---|---|---|---|
| **`s.replace("110", "101")`** | **No** | O(n) | O(n) | Fails whenever `"110"` overlaps a `"111"` block (`"1110"`). |
| **While loop with `res += ...`** | Yes | O(n²) worst | O(n) | Repeated string concatenation can TLE at `n = 10^5`. |
| **While loop + `list` + `"".join()` (Optimal)** | **Yes** | **O(n)** | **O(n)** | Exact spec fidelity and guaranteed linear runtime. |

## Real Assessment Q2 — Consensus Orchestrator for Translation Quality Review

**Category:** AI Agent Build / Deterministic Orchestration · **Difficulty:** Medium · **Evidence [CR]:** Accenture (AI Engineer OA)

**Problem Summary**
Implement `review(request, panel)` to run a 5-rater localization panel sequentially (`accuracy`, `terminology`, `fluency`, `localization_policy`, `formatting`), combine their votes deterministically, and return a structured `ReviewResult` with an exact attempt-level `trace`.

**Rules (Evaluated in Strict Order)**
1. **Invalid Up Front:** Missing/blank `segment.id` or `segment.text` → `status="invalid"`, `reason=None`, `vote_summary=None`, `trace=[]` (0 raters called).
2. **Bad Config / Unexpected Error (Never Raise):** Invalid config bounds or unexpected internal errors → `status="errored"`, with `error` message.
3. **Retries:** `RaterUnavailableError` or `None` response is retried up to `max_retries` times (`1 + max_retries` total attempts). Every attempt adds one trace entry (`attempt` starting at `1`). Other exceptions or malformed responses are **not** retried.
4. **Confidence Filter:** Only votes with `confidence >= min_confidence` count; lower-confidence votes increment `invalid` in `vote_summary`.
5. **Hard Issue Short-Circuit:** If a counted vote has `hard_issue=True`, return `status="blocked"`, `reason="HARD_ISSUE"` **immediately** without calling remaining raters.
6. **Quorum:** After all raters finish: `approvals >= approve_quorum` → `approved / QUORUM_APPROVE`; `rejections >= reject_quorum` → `rejected / QUORUM_REJECT`. If **both** quorums are met → escalate as `SPLIT_NO_WINNER`.
7. **Escalation:** If no quorum: `responders < min_responders` → `escalated / INSUFFICIENT_VOTERS`, else `escalated / SPLIT_NO_WINNER`.

```python
from __future__ import annotations
from dataclasses import dataclass, field, asdict
from typing import Any, Callable

class RaterUnavailableError(Exception):
    pass

@dataclass
class RaterResponse:
    vote: str            # 'approve' | 'reject' | 'abstain'
    confidence: float    # 0.0 .. 1.0
    hard_issue: bool = False

@dataclass
class MockRater:
    step: str
    behavior: Callable[[dict[str, Any]], RaterResponse | None]
    calls: int = 0

    def rate(self, segment: dict[str, Any]) -> RaterResponse | None:
        self.calls += 1
        return self.behavior(segment)

@dataclass
class ReviewResult:
    segment_id: str | None
    status: str
    reason: str | None = None
    vote_summary: dict[str, int] | None = None
    error: str | None = None
    trace: list[dict[str, Any]] = field(default_factory=list)

    def to_dict(self) -> dict[str, Any]:
        return asdict(self)

def review(request: dict[str, Any], panel: list[MockRater]) -> ReviewResult:
    """Deterministic consensus orchestrator following all 7 rules in strict order."""
    try:
        segment = request.get("segment") or {}
        seg_id = segment.get("id")
        seg_text = segment.get("text")

        # Rule 1: Validate segment up front
        if not isinstance(seg_id, str) or not seg_id.strip() or not isinstance(seg_text, str) or not seg_text.strip():
            return ReviewResult(
                segment_id=seg_id if isinstance(seg_id, str) else None,
                status="invalid",
                reason=None,
                vote_summary=None,
                error="INVALID_REQUEST: segment.id and segment.text are required",
                trace=[],
            )

        # Rule 2: Validate config
        cfg = request.get("config") or {}
        min_conf = cfg.get("min_confidence", 0.7)
        approve_q = cfg.get("approve_quorum", 3)
        reject_q = cfg.get("reject_quorum", 3)
        max_retries = cfg.get("max_retries", 1)
        min_resp = cfg.get("min_responders", 3)

        if any(isinstance(v, bool) for v in (min_conf, approve_q, reject_q, max_retries, min_resp)):
            raise ValueError("Boolean values are not valid numeric config fields")
        if not (0.0 <= min_conf <= 1.0) or not (0 <= max_retries <= 5):
            raise ValueError("Config out of bounds")

        summary = {"approvals": 0, "rejections": 0, "abstentions": 0, "invalid": 0, "failed": 0, "responders": 0}
        trace: list[dict[str, Any]] = []

        # Rule 3, 4, 5: Process panel sequentially with retries, confidence filter, and hard-issue short-circuit
        for rater in panel:
            final_resp: RaterResponse | None = None
            for attempt in range(1, max_retries + 2):  # 1 + max_retries attempts
                try:
                    resp = rater.rate(segment)
                except RaterUnavailableError:
                    trace.append({"step": rater.step, "attempt": attempt, "status": "unavailable"})
                    continue
                except Exception as exc:
                    trace.append({"step": rater.step, "attempt": attempt, "status": "error", "error": str(exc)})
                    break  # Non-retryable exception

                if resp is None:
                    trace.append({"step": rater.step, "attempt": attempt, "status": "missing"})
                    continue

                entry: dict[str, Any] = {
                    "step": rater.step,
                    "attempt": attempt,
                    "status": "success",
                    "vote": resp.vote,
                    "confidence": resp.confidence,
                }
                if resp.hard_issue:
                    entry["hard_issue"] = True
                trace.append(entry)
                final_resp = resp
                break

            if final_resp is None:
                summary["failed"] += 1
                continue

            # Rule 4: Confidence filter (must pass >= min_confidence to count)
            if final_resp.confidence < min_conf or final_resp.vote not in ("approve", "reject", "abstain"):
                summary["invalid"] += 1
                continue

            summary["responders"] += 1
            if final_resp.vote == "approve":
                summary["approvals"] += 1
            elif final_resp.vote == "reject":
                summary["rejections"] += 1
            else:
                summary["abstentions"] += 1

            # Rule 5: Hard-issue immediate block
            if final_resp.hard_issue:
                return ReviewResult(seg_id, "blocked", "HARD_ISSUE", summary, None, trace)

        # Rule 6 & 7: Quorum & Escalation
        app_met = summary["approvals"] >= approve_q
        rej_met = summary["rejections"] >= reject_q

        if app_met and rej_met:
            return ReviewResult(seg_id, "escalated", "SPLIT_NO_WINNER", summary, None, trace)
        if app_met:
            return ReviewResult(seg_id, "approved", "QUORUM_APPROVE", summary, None, trace)
        if rej_met:
            return ReviewResult(seg_id, "rejected", "QUORUM_REJECT", summary, None, trace)
        if summary["responders"] < min_resp:
            return ReviewResult(seg_id, "escalated", "INSUFFICIENT_VOTERS", summary, None, trace)
        return ReviewResult(seg_id, "escalated", "SPLIT_NO_WINNER", summary, None, trace)

    except Exception as exc:
        return ReviewResult(None, "errored", None, None, str(exc), [])

# --- Smoke Tests ---
base_req = {
    "segment": {"id": "seg-001", "text": "Speichern Sie {count} Artikel"},
    "config": {"min_confidence": 0.7, "approve_quorum": 3, "reject_quorum": 3, "max_retries": 1, "min_responders": 3},
}

# Test 1: Hard issue on rater 2 short-circuits remaining 3 raters
panel_hard = [
    MockRater("accuracy", lambda s: RaterResponse("approve", 0.93)),
    MockRater("terminology", lambda s: RaterResponse("reject", 0.92, hard_issue=True)),
    MockRater("fluency", lambda s: RaterResponse("approve", 0.90)),
]
res1 = review(base_req, panel_hard)
print("Test 1 (Hard issue):", res1.status, res1.reason, "| fluency calls:", panel_hard[2].calls)

# Test 2: Invalid blank text
res2 = review({"segment": {"id": "seg-002", "text": "   "}, "config": {}}, panel_hard)
print("Test 2 (Invalid):   ", res2.status, res2.reason, "| trace len:", len(res2.trace))
```

### Explanation & Complexity Analysis (Real Assessment Q2)

**How We Achieved This Solution**
- **Strict State-Machine Ordering:** Every rule is evaluated in the exact priority order specified so edge cases never collide (e.g., blank input exits before calling any rater; `confidence < min_confidence` is filtered before checking `hard_issue`; `app_met and rej_met` is checked before `if app_met`).
- **Safe Dataclass Defaults:** Using `field(default_factory=list)` for `trace` avoids the mutable default argument bug (Rank 10) leaking traces across calls.
- **Never-Raise Boundary:** Wrapping the orchestrator in a top-level `try/except` guarantees tests always receive a `ReviewResult` object with a `.status` attribute (preventing `'NoneType' object has no attribute 'status'`).

**How Time & Space Complexity Work**
- **Time — O(P · (1 + R)):** Where `P` is the number of raters in the panel (`5`) and `R = max_retries` (`<= 5`). Each rater runs at most `1 + R` times sequentially.
- **Space — O(P · (1 + R)):** Proportional to the number of trace entries logged (one entry per attempt).

**Design Trade-offs (Sequential vs. Async)**

| Design Choice | Pros | Cons / Why Spec Chose It |
|---|---|---|
| **Sequential Panel Execution (Spec)** | Deterministic trace order; hard-issue short-circuit saves downstream LLM/API calls. | Higher wall-clock latency when all 5 raters run. |
| **Parallel (`asyncio.gather`)** | Lower latency (bounded by slowest rater). | Non-deterministic trace order; wastes calls on raters after a hard issue. |
