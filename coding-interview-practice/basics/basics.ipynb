### 1. Dictionary (`dict`)

A dictionary stores data in **key-value pairs**. It is mutable and maintains insertion order (Python 3.7+).

**Insert / Set Values:**

* Use bracket notation `d[key] = value` to add a new key or overwrite an existing one.
* Use `.update({...})` to insert or merge multiple key-value pairs at once.
* Use `.setdefault(key, default)` to insert a key with a default value *only* if the key does not already exist.

```python
# Create
user = {"name": "Alice"}

# Insert / Set
user["age"] = 30                 # Adds new key 'age'
user["name"] = "Alice Smith"     # Overwrites existing key 'name'
user.update({"role": "Admin", "city": "NYC"})  # Adds multiple pairs
user.setdefault("country", "USA")              # Adds 'country' only if missing

```

**How to Iterate:**

```python
# 1. Iterate over keys (default behavior)
for key in user:
    print(key, user[key])

# 2. Iterate over values
for value in user.values():
    print(value)

# 3. Iterate over key-value pairs (Most Common & Recommended)
for key, value in user.items():
    print(f"{key}: {value}")

```

### 2. List (`list`)

A list is an **ordered, mutable sequence** of elements.

**Insert / Set Values:**

* `.append(item)`: Adds an item to the **end** of the list.
* `.insert(index, item)`: Adds an item at a **specific position** (shifts remaining items right).
* `.extend(iterable)`: Appends multiple elements from another collection.
* `lst[index] = value`: **Overwrites** the value at an existing index.

```python
fruits = ["apple", "banana"]

# Insert / Modify
fruits.append("orange")         # ['apple', 'banana', 'orange']
fruits.insert(1, "mango")       # ['apple', 'mango', 'banana', 'orange']
fruits.extend(["grape", "kiwi"]) # Appends multiple items
fruits[0] = "cherry"            # Overwrites index 0 -> ['cherry', ...]

```

**How to Iterate:**

```python
# 1. Direct value iteration
for fruit in fruits:
    print(fruit)

# 2. With index and value using enumerate()
for index, fruit in enumerate(fruits):
    print(f"Index {index}: {fruit}")

```

### 3. Set (`set`)

A set is an **unordered collection of unique, hashable elements**.

**Insert / Set Values:**

* `.add(item)`: Adds a single unique element.
* `.update(iterable)`: Adds multiple elements from another collection (duplicates are automatically ignored).
* *(Note: You cannot set values by index because sets are unordered and unindexed).*

```python
tags = {"python", "ai"}

# Insert
tags.add("machine-learning")     # Adds single item
tags.add("python")               # Ignored (already exists)
tags.update(["cloud", "docker"]) # Adds multiple items

```

**How to Iterate:**

```python
# Iterates through items (order is not guaranteed)
for tag in tags:
    print(tag)

```

### 4. Tuple (`tuple`)

A tuple is an **ordered, immutable sequence**. Once created, its contents cannot be modified, added, or reassigned.

**Insert / Set Values:**

* You **cannot** insert or set items in-place (e.g., `t[0] = 5` raises a `TypeError`).
* To "modify" or "add" to a tuple, you must **create a new tuple** by concatenating or converting to a list.

```python
coords = (10, 20)

# Concatenate to create a new tuple
coords = coords + (30,)  # Result: (10, 20, 30) (Note the trailing comma)

# Or convert to list, modify, and convert back
temp = list(coords)
temp.append(40)
coords = tuple(temp)     # Result: (10, 20, 30, 40)

```

**How to Iterate:**

```python
# 1. Direct value iteration
for val in coords:
    print(val)

# 2. With index and value
for idx, val in enumerate(coords):
    print(f"{idx} -> {val}")

```

### Quick Comparison Matrix

| Data Structure | Mutable? | Ordered? | Allows Duplicates? | Primary Insert Method |
| --- | --- | --- | --- | --- |
| **`dict`** | Yes | Yes (Python 3.7+) | Keys: No, Values: Yes | `d[key] = val`, `.update()` |
| **`list`** | Yes | Yes | Yes | `.append()`, `.insert()`, `l[i] = v` |
| **`set`** | Yes | No | **No** | `.add()`, `.update()` |
| **`tuple`** | **No** (Immutable) | Yes | Yes | `t = t + (new_item,)` |

````python
user={"name": "Alice"}

user.update({"role": "Admin", "city": "NYC"})

print(user)

user.setdefault("country", "USA")

print(user)


for key, value in user.items():
  print(f"{key}: {value}")

for value in user.values():
  print(value)

for key in user:
  print(key)
````
